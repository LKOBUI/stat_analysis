```python
import numpy as np
from scipy.stats import norm

# Define the prior distribution
prior_mean = 0
prior_std = 1

# Define the likelihood function
def likelihood(data, mu):
    return norm.pdf(data, loc=mu, scale=1)

# Generate some sample data
data = np.array([1, 2, 3, 4, 5])

# Update the prior distribution using Bayes' theorem
posterior_mean = (prior_mean + np.sum(data)) / (1 + len(data))
posterior_std = np.sqrt(1 / (1 + len(data)))

# Print the posterior distribution
print("Posterior mean:", posterior_mean)
print("Posterior standard deviation:", posterior_std)
```
