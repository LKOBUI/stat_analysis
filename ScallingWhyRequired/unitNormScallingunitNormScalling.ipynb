```python
path = os.path.join(os.getcwd(), 'DataSet', 'AcetyleneData_10_3.txt')
df = pd.read_csv(path)
# print(df)
df.columns = ['P','T','H','C']
```

```python
def norm_scalling(x_array_1d):
    # formula PG113,book Eqn 3.55,3.56
    xij   = x_array_1d
    xjBar = np.mean(xij)
    sjSqr = np.sum(np.square(xij - xjBar))/(np.size(x_array_1d)-1)
    sj = np.sqrt(sjSqr)
    zij = (xij - xjBar)/sj
    #print('sj: ',sj,' xhBar: ',xjBar)
    #print(zij)
    return zij
```

```python
norm_scalling(df['T'].to_numpy())
norm_scalling(df['H'].to_numpy())
norm_scalling(df['C'].to_numpy())
```
