```python
# Method 1: Standrizations according book method


def applyStandrizations(numpy_1D_array):
    MEAN = np.mean(numpy_1D_array)
    # Caculate corrected sum of sqaure i.e. Sxx
    fx1 = np.sum(np.square(numpy_1D_array))
    fx2 = np.square(np.sum(numpy_1D_array)) / np.size(numpy_1D_array)
    Sxx = fx1 - fx2
    sqrt_of_corrected_sum_sqr = np.sqrt(Sxx)
    centring = numpy_1D_array - MEAN
    corrected_sum_sqr = centring / sqrt_of_corrected_sum_sqr
    return corrected_sum_sqr
```

```python
def Sxy(x, y):
    gp = np.multiply(x, y)
    fx = np.mean(np.multiply(x, y))
    Sxy = np.sum(np.multiply(x, y)) - (np.multiply(np.sum(x), np.sum(y)) / np.size(x))
    ret = (gp - fx) / np.sqrt(Sxy)
    print("=================== ", ret)
    return (gp - fx) / np.sqrt(Sxy)
```
