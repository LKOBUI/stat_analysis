### FW Selections Steps:

- Start with no predictors (just the intercept).
- Iteratively add polynomial terms (e.g., x, x^2, x^3, \ldots) one at a time.
- At each step, choose the term that improves model performance the most (based on metrics like adjusted R^2, 
  AIC, BIC, or cross-validation error).
- Stop when adding more terms doesn’t significantly improve the model


#### Foward selections in linear regression

```python
def forward_selection(data, target, candidate_features):
    selected = []
    remaining = list(candidate_features)
    current_score, best_new_score = float('inf'), float('inf')

    while remaining:
        scores_with_candidates = []
        for candidate in remaining:
            formula = "{} ~ {}".format(target, ' + '.join(selected + [candidate]))            
            model = sm.OLS.from_formula(formula, data).fit()
            aic = model.aic
            print(f"{formula}, aic score: {aic}")
            scores_with_candidates.append((aic, candidate))

        scores_with_candidates.sort()
        best_new_score, best_candidate = scores_with_candidates[0]

        if best_new_score < current_score:
            remaining.remove(best_candidate)
            selected.append(best_candidate)
            current_score = best_new_score
        else:
            break

    formula = "{} ~ {}".format(target, ' + '.join(selected))
    final_model = sm.OLS.from_formula(formula, data).fit()
    return final_model
```

#### Backward elimination in Linear regressions:

- Start with all predictors in the model.
- Fit the full model and **calculate p-values for each predictor**.
- Identify the **predictor with the highest p-value** (**least statistically significant**).
- If that p-value is **above a chosen threshold** (commonly 0.05), **remove that predictor**.
- Refit the model without the removed predictor.
- Repeat steps 3–5 until all remaining predictors have p-values below the threshold.


```python
def backward_elimination(X, y, significance_level=0.05):
    num_vars = X.shape[1]
    for i in range(num_vars):
        regressor_OLS = sm.OLS(y, X).fit()  # Fit model using Ordinary Least Squares (OLS)
        max_p_value = max(regressor_OLS.pvalues)  # Get the highest p-value
        if max_p_value > significance_level:  # If p-value is greater than the threshold, remove that predictor
            max_p_index = np.argmax(regressor_OLS.pvalues)
            X = np.delete(X, max_p_index, axis=1)
        else:
            break  # Stop when all p-values are below the significance level
    return X, regressor_OLS
```

### Backward elimination in polynomial regression

In polynomial regression, your predictors include:
- Linear terms: x
- Higher-order terms: x^2, x^3, \ldots, x^d
- Possibly interaction terms: x_1x_2, x_1^2x_2, \ldots

Backward elimination helps you remove terms that don’t significantly improve the model, based on statistical tests (usually p-values).


```python
X = np.random.rand(100, 1)
y = 3 * X.squeeze()**2 + 2 * X.squeeze() + np.random.randn(100)

# Generate polynomial features
poly = PolynomialFeatures(degree=4, include_bias=False)
X_poly = poly.fit_transform(X)

# Convert to DataFrame for easier column dropping
X_df = pd.DataFrame(X_poly, columns=poly.get_feature_names_out(['x']))
X_df = sm.add_constant(X_df)

# Backward elimination loop
while True:
    model = sm.OLS(y, X_df).fit()
    p_vals = model.pvalues.drop('const')
    max_p = p_vals.max()
    if max_p > 0.05:
        worst_term = p_vals.idxmax()
        X_df = X_df.drop(columns=[worst_term])
    else:
        break

print(model.summary())
```
