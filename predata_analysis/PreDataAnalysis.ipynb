#### 1. loc Method

**Note**: Loc method you use when you want to oprations on row and row with columns.
      But when you want to oprations on columns only, you can use directly dataframe.

df.loc[row_indexer, column_indexer]


- row_indexer: Label(s) of the row(s) you want to select.
- column_indexer: Label(s) of the column(s) you want to select.

Common Use Cases
1. Select a single row by label
- df.loc['row_label']


2. Select multiple rows
- df.loc[['row1', 'row2']]


3. Select a range of rows
- df.loc['row1':'row5']


4. Select specific columns
- df.loc[:, ['col1', 'col2']]
- df[['col1', 'col2']]


5. Select rows and columns together
- df.loc['row1', 'col1']           # Single value
- df.loc['row1':'row3', ['col1']]  # Subset


6. Conditional selection
- df.loc[df['col1'] > 100]
- df[df['col]>5]


7. Assigning values
- df.loc['row1', 'col1'] = 999

#### iloc Method

- **df[val]:**  Select single column or sequence of columns from the DataFrame; special case
conveniences: boolean array (filter rows), slice (slice rows), or boolean DataFrame
(set values based on some criterion)
- **df.loc[val]** : Selects single row or subset of rows from the DataFrame by label
- **df.loc[:, val]:** Selects single column or subset of columns by label
- **df.loc[val1, val2]:** Select both rows and columns by label
- **df.iloc[where]:** Selects single row or subset of rows from the DataFrame by integer position
- **df.iloc[:, where]:** Selects single column or subset of columns by integer position
- **df.iloc[where_i, where_j]:** Select both rows and columns by integer position

### Map Functions

In pandas, .map() is a **Series-level method** used to apply a function or mapping to each element in a Series. It’s ideal for element-wise transformations, especially when you want to replace values or apply custom logic.

Syntax:
Series.map(arg, na_action=None)

1. s = pd.Series([1, 2, 3])
   - s.map(lambda x: x ** 2)

2. Mapping your catergorical variable
-  df = pd.DataFrame({'grade': ['A', 'B', 'C']})
-  df['grade'].map({'A': 4, 'B': 3, 'C': 2})

3. Conditionals mappping:

   def categorize(x):

    return 'high' if x > 70 else 'low' 

   df['score'].map(categorize)








```python
# Sample DataFrame
df = pd.DataFrame({
    'A': [1.2345, 5.6789],
    'B': [9.1011, 2.3456]
})

# map on entire dataframe rounding to every cell
rounded_df = df.map(lambda x: round(x, 2))
print(rounded_df)
```

### Unique Values, Value Counts, and Membership

In pandas, .unique() is a Series method that returns the distinct values in that Series — supe


```python
df = pd.DataFrame({
    'color': ['red', 'blue', 'red', 'green', 'blue']
})

# Get unique colors
unique_colors = df['color'].unique()
print(f"unique values are : {unique_colors}")
print(df['color'].is_unique)
print('find duplicate')
df[df['color'].duplicated() == True]
print('Count values: ')
print(df['color'].value_counts())
```

### Handling Missing Data



```python
df = pd.DataFrame({
    'name': ['Alice', 'Bob', None],
    'score': [85, None, 90]
})

print(df.isnull())

# 1. Filter rows with missing data

df[df['score'].isnull()]

# 2. Count missing values

df.isnull().sum()

```

### Data Cleaning

```python
print('drop all nan value')
print(df.dropna())
print('Drop drom columns')
print(df['name'].dropna())

# Passing how='all' will only drop rows that are all NA:
df = pd.DataFrame([[1., 6.5, 3.], [1., None, None],
                  [None, None, None], [None, 6.5, 3.]])
df[4] = None
print('Drop rows contains all NA')
print(df.dropna(how='all'))

# To drop columns in the same way, pass axis=1:
print('To drop columns contains all none')
df.dropna(axis=1, how='all')
```

### Filling In Missing Data

```python
df = pd.DataFrame(np.random.randn(6, 3))
df.iloc[2:, 1] = None
df.iloc[4:, 2] = None
print('Data:')
print(df)
# Fill with 0

df.fillna(0)

# Fill with Mean

df.fillna(df.mean())
```

### Remove Duplicate

In pandas, drop_duplicates() is a powerful method used to remove duplicate rows from a DataFrame — either across all columns or based on specific ones.


```python
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Alice', 'David'],
    'score': [90, 85, 90, 88]
})

# Remove full-row duplicates
df_cleaned = df.drop_duplicates()

# Remove Duplicates Based on Specific Column(s)
df.drop_duplicates(subset='name', keep='first')

```

### pivot() in Pandas: What It Does

The pivot() method reshapes a DataFrame by turning unique values from one column into columns, using another column for row labels, and a third for values.

- DataFrame.pivot(index=None, columns=None, values=None)


```python
df = pd.DataFrame({
    'Date': ['2023-01', '2023-01', '2023-02', '2023-02'],
    'Product': ['A', 'B', 'A', 'B'],
    'Sales': [100, 150, 120, 160]
})

pivoted = df.pivot(index='Date', columns='Product', values='Sales')
print(pivoted)
```
