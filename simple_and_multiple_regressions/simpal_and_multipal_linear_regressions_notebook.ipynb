
#### **Sum of Squares and Parameter Estimation**

* **Sum of Squares for $X$ ($S_{xx}$):**
  $$S_{xx} = \sum x_i^2 = \sum(x_i - \bar{x})^2$$
  *(where $\bar{x}$ represents the mean of $x$)*

* **Sum of Product for $X$ and $Y$ ($S_{xy}$):**
  $$S_{xy} = \sum y_i x_i = \sum y_i(x_i - \bar{x})$$

* **Slope Estimator ($\hat{\beta}_1$):**
  $$\hat{\beta}_1 = \left(\frac{S_{xy}}{S_{xx}}\right)$$

#### **Residuals and Model Adequacy**

* **Residual Error ($e_i$):**
  $$e_i = (y_i - \hat{y}_i) = (y_i - (\hat{\beta}_0 + \hat{\beta}_1 x_i))$$
  * **$e_i$**: Residual (used to check model adequacy)
  * **$y_i$**: Actual value
  * **$(\hat{\beta}_0 + \hat{\beta}_1 x_i)$**: Predicted value
  * **$\hat{\beta}_1$**: Least squares estimator (Unbiased)

#### **Least Squares Estimator Expansion**

$$\beta = \frac{S_{xy}}{S_{xx}} = \frac{\sum y_i(x_i - \bar{x})}{\sum(x_i - \bar{x})^2}$$

* Note on the denominator expansion:
  $$\sum(x_i - \bar{x})^2 = \sum x_i^2 - \frac{(\sum x_i)^2}{n}$$

#### **BLUE Property (Gauss-Markov Theorem)**

* Least squares estimator $\rightarrow$ **Best Linear Unbiased Estimator (BLUE)**
* Here, **"Best"** means achieving **minimum variance** among all linear unbiased estimators.

* **Question:** What is the variance of the estimator if it is the best estimator?
* **Answer:** 
  $$\text{Var}(\hat{\beta}_1) = \frac{\sigma^2}{S_{xx}}$$
  $$\text{Var}(\hat{\beta}_0) = \sigma^2 \left( \frac{1}{n} + \frac{\bar{x}^2}{S_{xx}} \right)$$


#### **Properties of Least Squares Residuals and Variance Estimation**

* **Sum of Residuals is Zero:**
  $$\sum e_i = 0$$

* **Orthogonality of Residuals and Regressors:**
  $$\sum e_i x_i = 0$$
  * **$e_i$**: Residual
  * **$x_i$**: Regressor variable

* **Orthogonality of Residuals and Fitted Values:**
  $$\sum e_i \hat{y}_i = 0$$
  * **$\hat{y}_i$**: Fitted value

* **Sum of Observed Values Equals Sum of Fitted Values:**
  $$\sum y_i = \sum \hat{y}_i$$
  * **$y_i$**: Observed value
  * **$\hat{y}_i$**: Fitted value

* **Centroid Passage Property:**
  * The regression line determined by estimators $\hat{\beta}_0, \hat{\beta}_1$ always passes through the centroid of the data $(\bar{x}, \bar{y})$.

* **Estimation of Error Variance ($\sigma^2$):**
  * **Question:** Why is the estimation of $\sigma^2$ necessary?
  * **Answer:** It is required to conduct hypothesis testing and construct confidence intervals.

* **Residual Sum of Squares ($SS_{\text{res}}$):**
  * Denotes the residual or error sum of squares.
  * Degrees of Freedom ($\text{dof}$) $\Rightarrow n - 2$
  * *Reasoning for $n - 2$:* Two degrees of freedom are lost because two parameters ($\hat{\beta}_0$ and $\hat{\beta}_1$) must be estimated first from the data.

* **Unbiased Estimator of Variance ($\hat{\sigma}^2$):**
  $$\hat{\sigma}^2 = \left(\frac{SS_{\text{res}}}{n - 2}\right) = MS_{\text{res}}$$
  * **$MS_{\text{res}}$**: Mean Square Residual


#### **Model Dependent Estimation of $\sigma^2$ and Hypothesis Testing**

* **Unbiased Estimator of Variance:**
  $$\hat{\sigma}^2 = \left(\frac{SS_{\text{res}}}{n - 2}\right) = MS_{\text{res}}$$

* **Standard Error of Regression:**
  * $\hat{\sigma}$ or $\sqrt{MS_{\text{res}}}$ represents the standard error of regression.

* **Hypothesis Testing for Slope and Intercept:**
  * **Null Hypothesis ($H_0$):** $\beta_1 = \beta_{1,0}$ (or general parameter $\beta = \beta_x$)
  * **Alternative Hypothesis ($H_1$):** $\beta_1 \neq \beta_{1,0}$ (or general parameter $\beta \neq \beta_x$)

* **Test Statistic Calculation ($t_0$):**
  $$t_0 = \frac{\hat{\beta}_1 - \beta_{1,0}}{\sqrt{\frac{MS_{\text{res}}}{S_{xx}}}}$$

* **Degrees of Freedom ($\text{dof}$):**
  * The degrees of freedom associated with the test statistic $t_0$ is equal to the $\text{dof}$ of $MS_{\text{res}}$, which is $n - 2$.

* **Decision Rule and Critical Value Comparison:**
  * Compare $|t_0|$ with the upper $\alpha/2$ percentage point of the $t_{n-2}$ distribution.
  * Rejection Condition:
    $$|t_0| > t_{\alpha/2, n-2}$$
  * If this condition is true $\Rightarrow$ **Reject the Null Hypothesis**.
  * Significance Level ($\alpha$): Typically set to $0.05$ (p-value threshold).

#### **Testing Significance of Regression**

* **Hypothesis Framework:**
  * **Null Hypothesis ($H_0$):** $\beta_1 = 0$
  * **Alternative Hypothesis ($H_1$):** $\beta_1 \neq 0$
  * *Note:* This represents the hypothesis related to the overall significance of regression.

* **Implication of Failing to Reject $H_0$:**
  * Failing to reject the null hypothesis implies that there is no linear relationship between $x$ and $y$.
  * **$y$**: dependent variable
  * **$x$**: independent variable

* **Scenario A: $H_0$ Not Rejected ($p \geq \alpha$)**
  * Graphical indicators show a flat regression line with a slope $\beta_1 = 0$.
  * The distribution of data points results in favor of $H_0$.
  * **Conclusion:** No relationship exists between the independent variable $x$ and dependent variable $y$.

* **Scenario B: $H_0$ Rejected ($p \leq \alpha$)**
  * Graphical indicators show a distinct upward or downward sloped regression line where $\beta_1 \neq 0$.
  * The distribution of data points results in favor of $H_1$.
  * **Conclusion:** The null hypothesis $H_0 \Rightarrow \beta_1 = 0$ is rejected, confirming a significant relationship.

* **Significance Metric:**
  * $\alpha = 0.05 \rightarrow 5\%$ significance level.

#### **Analysis of Variance and Structural Test Statistics**

* **General Test Statistic ($t_0$):**
  $$t_0 = \frac{\hat{\beta}_1 - \beta_{1x}}{\text{SE}(\hat{\beta}_1)}$$

* **Scenario A: $\beta_{1x} = 0$ (Significance Test)**
  * **Null Hypothesis ($H_0$):** $\hat{\beta}_1 = \beta_{1x}$
  * **Alternative Hypothesis ($H_1$):** $\hat{\beta}_1 \neq \beta_{1x}$
  * *Purpose:* Determines whether an operational relationship exists between variables $x$ and $y$.

* **Scenario B: $\beta_{1x} \neq 0$ (Equivalence Test)**
  * **Null Hypothesis ($H_0$):** $\hat{\beta}_1 = \beta_{1x}$
  * **Alternative Hypothesis ($H_1$):** $\hat{\beta}_1 \neq \beta_{1x}$
  * *Purpose:* Tests if the calculated estimator $\hat{\beta}_1$ is equivalent to a specific non-zero value $\beta_{1x}$.

* **Confidence Bounds Interval:**
  * Uses critical threshold mapping values: $100(1-\alpha)\%$ with critical multiplier $t_{\alpha/2, n-2}$.

#### **Analysis of Variance (ANOVA) Framework**

* **Sum of Squares Partitioning Components:**
  * **$SS_{\text{res}}$**: Residual sum of squares $\Rightarrow (n-2)MS_{\text{res}} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$
  * **$SSR$**: Regressor (Regression) sum of squares $\Rightarrow (\hat{\beta}_1 S_{xy})$
  * **$SS_T$**: Total sum of squares

* **Fundamental ANOVA Equation:**
  $$SS_T = SS_{\text{res}} + SSR$$

#### **Partitioning Degrees of Freedom in ANOVA**

* **Total Sum of Squares ($SS_T$):**
  $$SS_T = SSR + SS_{\text{res}}$$
  * **Degrees of Freedom ($df_T$):** $n - 1$
  * *Reasoning:* The subtraction of $1$ indicates that one degree of freedom is lost due to the constraint equation $\sum_{i=1}^{n}(y_i - \bar{y}) = 0$, which represents the deviation of $y$ from its mean value.

* **Regression Sum of Squares ($SSR$):**
  * **Degrees of Freedom ($df_R$):** $1$
  * *Reasoning:* $SSR$ is completely determined by a single slope parameter estimate, namely $\hat{\beta}_1$.

* **Residual Sum of Squares ($SS_{\text{res}}$):**
  * **Degrees of Freedom ($df_{SS_{\text{res}}}$):** $n - 2$
  * *Reasoning:* Two structural constraints are imposed on the deviations $(y_i - \hat{y}_i)$ as a direct result of estimating the parameters $\hat{\beta}_0$ and $\hat{\beta}_1$.

#### **F-Statistic Derivation and Distributional Theory**

* **Degrees of Freedom Vector Equation:**
  $$df_T = df_R + df_{\text{res}}$$
  $$(n - 1) = (1) + (n - 2) \leftarrow \text{dof}$$

#### **Theoretical Requirements for ANOVA F-Test Execution**

* **Distribution of Regression Sum of Squares ($SSR$):**
  * If the Null Hypothesis $H_0: \beta_1 = 0$ is true, then $SSR$ follows a $\chi^2$ distribution with $1$ degree of freedom.

* **Distribution of Residual Sum of Squares ($SS_{\text{res}}$):**
  * Given $SS_{\text{res}} = (n-2)MS_{\text{res}}$, this term structurally follows a $\chi^2_{n-2}$ distribution.

* **Independence Condition:**
  * $SS_{\text{res}}$ and $SSR$ are statistically independent random variables.

* **F-Statistic Formulation ($F\text{-stat}$):**
  $$F\text{-stat} = \frac{\frac{SSR}{df_R}}{\frac{SS_{\text{res}}}{df_{\text{res}}}} = \frac{\frac{SSR}{1}}{\frac{SS_{\text{res}}}{n-2}} = \frac{MSR}{MS_{\text{res}}}$$
  * **$MSR$**: Mean Square Regression
  * **$MS_{\text{res}}$**: Mean Square Residual
  * This computed ratio follows an $F_{1, n-2}$ distribution, where $1$ is the $df$ of $SSR$ and $n-2$ is the $df$ of $SS_{\text{res}}$.

#### **F-Test Rejection Rules and Non-Central Distributions**

* **Rejection Decision Rule for $H_0: \beta_1 = 0$:**
  * The null hypothesis is rejected if the computed test statistic $F_0$ is large.
  * Rejection Condition:
    $$F_0 > F_{\alpha, 1, n-2}$$
  * $\Rightarrow$ **Reject the Null Hypothesis**
  * **$1$**: degrees of freedom for $SSR$
  * **$n - 2$**: degrees of freedom for $SS_{\text{res}}$

* **Distribution Properties When $F_0$ Is Large:**
  * When $H_0$ is false, the statistic follows a **Non-Central F-distribution** with $1$ and $n - 2$ degrees of freedom.
  * Non-centrality parameter calculation ($\lambda_{\text{noncentral}}$):
    $$\lambda_{\text{noncentral}} = \frac{\beta_1^2 S_{xx}}{\sigma^2}$$

#### **Structural Comparison of Hypothesis Tests**

* **t-test:**
  * Primarily used to evaluate one-sided (directional) alternative hypotheses.

* **F-test:**
  * Geometrically evaluates two-sided alternative hypotheses frameworks globally.

#### **Confidence Intervals and Interval Estimation of Mean Response**

* **Confidence Intervals ($\text{CI}$):**
  * Establishes confidence intervals for the parameters $\beta_0$, $\beta_1$, and variance $\sigma^2$.

#### **Interval Estimation of Mean Response**

* **Definition:**
  * Estimating the expected mean response value $E(y)$ corresponding to a specific, selected target value of $x$.

* **Practical Field Example:**
  * *Scenario:* Analyzing the mean shear strength of a propellant bond layer within a rocket motor system.
  * *Objective:* Estimating the mean response criteria $E(y | x_0)$.

* **Operational Target Context:**
  * Obtains the definitive target value of the expected mean response $y$ at a given localized coordinate point $x_0$.
  * **$x_0$**: Represents any valid input regressor value located strictly within the original bounding range of the training data vectors on $x$ used to fit the model.


#### **Prediction of New Observations and Coefficient of Determination ($R^2$)**

* **Prediction of New Observations:**
  * Establishes a prediction interval for a discrete future observation value ($y_0$).

#### **Coefficient of Determination ($R^2$)**

* **Mathematical Formula:**
  $$R^2 = \frac{SSR}{SS_T} = \frac{SS_T - SS_{\text{res}}}{SS_T} = 1 - \frac{SS_{\text{res}}}{SS_T}$$

* **Boundary Constraints and Proof:**
  * Since $0 \leq SS_{\text{res}} \leq SS_T$:
    * If $SS_{\text{res}} = 0 \Rightarrow R^2 = 1$
    * If $SS_{\text{res}} = SS_T \Rightarrow R^2 = 0$
  * Therefore, the value is strictly bounded:
    $$0 \leq R^2 \leq 1$$

* **Core Analytical Interpretation:**
  * **Question:** What specific insight does $R^2$ provide?
  * **Inherent Issue with $SS_T$:** $SS_T$ measures the total raw variability in $y$ without considering the direct impact of the regressor variable $x$.
  * **Role of $SS_{\text{res}}$:** Conversely, $SS_{\text{res}}$ measures the remaining unexplained variability in $y$ after accounting for the regression model effects.

#### **Proportion of Variation Explained and R² Characteristics**

* **Definition of $R^2$:**
  * It represents the proportion of variation in the dependent variable explained by the regressor variable $x$.

* **Practical Interpretation Example:**
  $$R^2 = \frac{SSR}{SS_T} = \frac{1527334.95}{1693737.6} = 0.9018$$
  * **Conclusion:** We can say that $90.18\%$ of the variability in strength is accounted for by the regression model.

#### **Critical Insights on R² Behavior**

* **Impact of Model Size:**
  * You can increase the $R^2$ value artificially simply by adding enough terms to the model.

* **Polynomial Fit Edge Case:**
  * For a model with no repeated points (meaning no multiple distinct $y$ values measured for the exact same $x$ value), a polynomial with a degree of $(n-1)$ will yield a perfect fit where $R^2 \approx 1$.

* **Repeated Points Limitation:**
  * If repeated experimental data points are present in the dataset, the maximum possible $R^2$ can never exactly equal $1$.

#### **Expected Value of $R^2$ and Range of Variability**

* **Note on Model Selection:**
  * Adding additional terms into a model simply to increase the $R^2$ value is not a good approach.

* **Expected Value of $R^2$ Equation:**
  $$E(R^2) = \frac{\hat{\beta}_1^2 SS_{xx}}{\hat{\beta}_1^2 SS_{xx} + \sigma^2}$$

* **Sum of Squares Expansion ($SS_{xx}$):**
  $$SS_{xx} = \sum_{i=1}^{n} x_i^2 - \frac{(\sum_{i=1}^{n} x_i)^2}{n}$$

* **Dependency on Regressor Range Variability:**
  * The magnitude of $R^2$ depends directly upon the range of variability in the independent regressor variable.
  * $R^2$ increases if the spatial spread of the $x$ values is increased.
  * $R^2$ decreases if the spatial spread of the $x$ values is decreased.
  * Therefore: $R^2 \uparrow(\downarrow) \rightarrow SS_{xx} \uparrow(\downarrow)$

#### **Core Misinterpretations of the Coefficient of Determination**

* $R^2$ does not measure the actual numerical magnitude of the regression slope.
* A large value of $R^2$ does not necessarily imply a steep slope configuration.
* $R^2$ does not measure the fundamental appropriateness of choosing a linear model over non-linear models.

#### **Predictive Limits of R² and Introduction to Maximum Likelihood**

* **Core Cautionary Rule:**
  * Even if $R^2$ is large, it does not necessarily imply that the regressor model will be an accurate predictor for future outcomes.

#### **Critical Diagnostic Considerations in Linear Regression**

* **Outliers**: Extreme data anomalies distorting baseline trends.
* **Multicollinearity**: High inter-correlations among separate independent channels.
* **Interpolation**: Making localized predictions strictly within original data bounds.
* **Extrapolation**: High-risk forecasting attempted outside observed sample ranges.

#### **Maximum Likelihood Estimation Framework**

* **Foundational Conceptual Origin:**
  * **Question:** Where does Maximum Likelihood Estimation ($\text{MLE}$) come from and why do we use it?
  * **Answer:** It acts very similarly to the ordinary least squares ($\text{OLS}$) estimator framework.
  * While the $\text{OLS}$ framework optimizes geometric parameters by minimizing squared spatial residuals, the $\text{MLE}$ approach performs a similar parameter tracking task using structural probability density calculations.
  * Therefore, $\text{MLE}$ is an alternative statistical estimator family used to estimate the linear regression parameters, namely: $\hat{\beta}_0, \hat{\beta}_1,$ and the error variance variance variable $\sigma^2$.

#### **Maximum Likelihood Conditions for Parameter Estimation**

* **Distribution of Residual Errors:**
  * In linear regression, we assume errors in the regressor model are normally and independently distributed:
    $$\epsilon_i \sim \text{NID}(0, \sigma^2)$$
  * **$0$**: Zero mean value parameter
  * **$\sigma^2$**: Constant variance parameter

* **Distribution of Dependent Variable ($y_i$):**
  * Under this normality condition, the observations $y_i$ become normally and independently distributed random variables.
  * Mean value parameter: $(\beta_0 + \beta_1 x_i)$
  * Variance parameter: $\sigma^2$

* **Likelihood Function for Normal Linear Regression ($L$):**
  $$L(y_i, x_i, \beta_0, \beta_1, \sigma) = \prod_{i=1}^{n} \frac{1}{\sqrt{2\pi\sigma^2}} e^{\left[ -\frac{1}{2\sigma^2}(y_i - \beta_0 - \beta_1 x_i)^2 \right]}$$

#### **Maximum Likelihood Estimators ($\text{MLE}$)**

* Solving the likelihood function optimization equations yields the definitive estimators:
  $$\tilde{\beta}_0 = \bar{y} - \tilde{\beta}_1 \bar{x}$$
  $$\tilde{\beta}_1 = \frac{\sum y_i(x_i - \bar{x})}{\sum(x_i - \bar{x})^2}$$
  $$\sigma^2 = \frac{\sum (y_i - \tilde{\beta}_0 - \tilde{\beta}_1 x_i)^2}{n}$$

#### **Maximum Likelihood vs. Least Squares and Multiple Regression Introduction**

* **Parameter Value Convergence Note:**
  * Maximum likelihood estimation ($\text{MLE}$) produces the exact same intercept and slope parameter values as the ordinary least squares ($\text{OLS}$) method.

* **Divergence Properties:**
  * **Question:** How does maximum likelihood differ from ordinary least squares?
  * **Answer:** Maximum likelihood estimators possess better asymptotic statistical properties compared to ordinary least squares ($\text{LS}$) in large samples.
  * *Research Core:* Investigate the specific conditions under which these advanced properties (such as asymptotic efficiency and consistency) hold compared to least squares.

#### **Multiple Linear Regression Models**

* **Contour Plots:**
  * Also referred to as a **level plot**.
  * It serves as a visual diagnostic tool for performing multivariate data analysis.
  * Enables the visualization of 3-D response surface topologies directly within a flat 2-D coordinate space.

#### **Multiple Linear Regression Mathematical Formulation**

* **General Multiple Regression Model Equation:**
  $$y_i = \beta_0 + \beta_1 x_{i1} + \beta_2 x_{i2} + \dots + \beta_k x_{ik} + \epsilon_i$$

* **Compact Summation Notation:**
  $$y_i = \beta_0 + \sum_{j=1}^{k} \beta_j x_{ij} + \epsilon_i$$
  * **$i$**: Index representing individual observations spanning from $1, 2, 3, \dots, n$.
  * **$x_{ij}$**: Explanatory matrix variables.
  * **$i$ (from $1$ to $n$)**: Denotes index row expansion operations.
  * **$j$ (from $1$ to $k$)**: Denotes index column expansion operations.

#### **Least Squares Minimization Objective Function**

* **Sum of Squared Residuals Function ($S$):**
  $$S(\beta_0, \beta_1, \dots, \beta_k) = \sum_{i=1}^{n} \epsilon_i^2$$
  $$S(\beta_0, \beta_1, \dots, \beta_k) = \sum_{i=1}^{n} \left( y_i - \beta_0 - \sum_{j=1}^{k} \beta_j x_{ij} \right)^2$$

#### **Matrix Notation Representation**

* **Linear Regression Matrix Framework:**
  * Alternatively, the complete system can be written compactly using standard matrix algebra notation:
  $$Y = X\beta + \epsilon$$


#### **Matrix Dimensions and Structural Components**

* **Design Matrix ($X$):**
  $$X = \begin{bmatrix} 
  1 & x_{11} & x_{12} & \dots & x_{1k} \\ 
  1 & x_{21} & x_{22} & \dots & x_{2k} \\ 
  1 & x_{31} & x_{32} & \dots & x_{3k} \\ 
  \vdots & \vdots & \vdots & & \vdots \\ 
  1 & x_{n1} & x_{n2} & \dots & x_{nk} 
  \end{bmatrix}$$
  * Dimensionality configuration: $n \times p$ matrix

* **Observation, Error, and Parameter Vectors:**
  $$y = \begin{bmatrix} y_1 \\ y_2 \\ y_3 \\ \vdots \\ y_n \end{bmatrix}, \quad \epsilon = \begin{bmatrix} \epsilon_1 \\ \epsilon_2 \\ \epsilon_3 \\ \vdots \\ \epsilon_n \end{bmatrix}, \quad \beta = \begin{bmatrix} \beta_0 \\ \beta_1 \\ \vdots \\ \beta_k \end{bmatrix}$$
  * **$y$**: $n \times 1$ observation vector
  * **$\epsilon$**: $n \times 1$ random error vector
  * **$\beta$**: $p \times 1$ parameter vector

* **Matrix Structural Properties:**
  * Bounding matrix size rule: $X \rightarrow n \times p$
  * **$p$**: Evaluates to $(k + 1)$
  * **$p$**: Represents the number of rows or the number of variables in columns.
  * **$k$**: Explanatory or independent variable count.

* **Vector Dimension Definitions:**
  * $y \rightarrow n \times 1$ vector
  * $\epsilon \rightarrow n \times 1$ vector
  * $\beta \rightarrow p \times 1$ vector

#### **Ordinary Least Squares Matrix Solutions and the Hat Matrix**

* **Least Squares Parameter Estimator Formula ($\hat{\beta}$):**
  $$\hat{\beta} = (X'X)^{-1}X'y$$
  * Derived directly from solving the linear normal equations system.

* **Inversion Existence Conditions:**
  * Requires that the inverse matrix calculation $(X'X)^{-1}$ exists mathematically.
  * This product matrix always exists and is invertible if the individual input regressor columns are strictly linearly independent.

* **Variable Definition Flags:**
  * **$y_i$**: Observed target data value.
  * **$\hat{y}_i$**: Fitted model prediction value.

#### **The Hat Matrix Architecture ($H$)**

* **Mathematical Derivation Matrix Sequence:**
  $$\hat{y} = X\hat{\beta} = X(X'X)^{-1}X'y = Hy$$

* **Hat Matrix Formula ($H$):**
  $$H = X(X'X)^{-1}X'$$
  * Structural size configuration: $n \times n$ matrix dimensional space.

* **Geometric Matrix Purpose:**
  * It maps the vector of observed values directly into a vector of fitted values.

* **Residual Vector Equation ($e_i$):**
  $$e_i = y_i - \hat{y}_i$$

#### **Residual Transformation Tensors and Step-by-Step Numerical Modeling**

* **Residual Matrix Formula Deconstruction:**
  $$e = y - X\hat{\beta} = y - Hy = (I - H)y$$
  * **$I$**: Identity matrix mapping layout.
  * **$H$**: Standard hat matrix mapping transformation.

#### **Procedural Operational Execution Workflow**

* **Target Structural Problem:**
  * Find the corresponding linear regression equation parameters matching:
    $$y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \epsilon$$

* **Step 1: Calculate Dot Product Transposition Matrices:**
  * Compute structural arrays $X'X$ and $X'y$.

* **Step 2: Solve Parameter Matrix Equation:**
  * Find parameter weight vector: $\hat{\beta} = (X'X)^{-1}X'y$

* **Step 3: Establish Definitive Output Prediction Line:**
  $$y = 2.341 + 1.615 x_1 + 0.0143 x_2$$

* **Step 4: Evaluate Residual Error Mean Boundaries:**
  * **$\epsilon$**: Normalized random error matrix channel.
  * Structural expectation balancing condition: $\sum \epsilon = 0$

#### **Covariance Properties of Estimated Weights ($\hat{\beta}$)**

* **Global Matrix Covariance Equation:**
  $$\text{Cov}(\hat{\beta}) = \sigma^2(X'X)^{-1}$$

* **Diagonal Variance Components Mapping:**
  * If matrix inverse array is assigned as: $C = (X'X)^{-1}$
  * Individual index parameter variance evaluates to:
    $$\text{Var}(\hat{\beta}_j) = \sigma^2 c_{jj}$$
  * Individual cross-parameter covariance evaluates to:
    $$\text{Cov}(\hat{\beta}_i, \hat{\beta}_j) = \sigma^2 c_{ij}$$

#### **Gauss-Markov Theorem and Residual Mean Square Estimation**

* **Gauss-Markov Fundamental Statement:**
  * The least-squares estimator $\hat{\beta}$ is the Best Linear Unbiased Estimator (BLUE) of $\beta$.

#### **Estimation of Error Variance ($\sigma^2$) via Residual Mean Square**

* **Unbiased Estimator of Variance Equation:**
  $$\hat{\sigma}^2 = MS_{\text{res}} = \left( \frac{SS_{\text{res}}}{n - p} \right) \quad \text{--- (1)}$$

* **Degrees of Freedom Framework for Residuals:**
  * Degrees of Freedom ($\text{DOF}$) for $SS_{\text{res}} \Rightarrow n - p$
  * **$p$**: Total parameter count evaluated as $(k + 1)$
  * **$k$**: Total number of independent estimator or regressor variables.

* **Question:** Which empirical model specification is considered correct or preferred?

  * Since $\sigma^2$ represents the statistical variance of the underlying error term, we would always prefer an operational model that yields a small overall **Residual Mean Square** ($MS_{\text{res}}$).

* **Matrix Representation of Residual Sum of Squares ($SS_{\text{res}}$):**

#### **Maximum Likelihood Estimation (MLE) Framework**

* **Structural Model Equation:**
  * When model error terms are normally and independently distributed, the matrix regression system is formulated as:
    $$Y = X\beta + \epsilon$$

* **Distributional Error Conditions:**
  * The error terms are assumed to be normally and independently distributed with a constant variance component $\sigma^2$.
  * Operational notation condition for normally distributed error:
    $$\epsilon \text{ is distributed as } N(0, \sigma^2)$$

* **Normal Probability Density Function for Errors:**
  * Given the structural normality assumptions, individual errors track along the density function:
    $$f(\epsilon_i) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{1}{2\sigma^2}\epsilon_i^2} \quad \text{--- (1)}$$

#### **Likelihood Maximization and Parameter Equivalence Proof**

* **Likelihood Product Function Formulation:**
  * Substituting the error term density function into the joint product equation:
    $$L(\epsilon, \beta, \sigma^2) = \prod_{i=1}^{n} f(\epsilon_i) = \frac{1}{(2\pi)^{n/2} \sigma^n} e^{-\frac{1}{2\sigma^2} \sum_{i=1}^{n} \epsilon_i^2}$$

* **Maximization Condition Link:**
  * For a fixed value of $\sigma$, the joint log-likelihood function is maximized precisely when the residual sum of squares vector product is minimized:
    $$\text{Minimize: } (y - X\beta)'(y - X\beta)$$

* **Estimator Equivalence Property:**
  * The maximum likelihood estimator of $\beta$ under normal error assumptions is completely equivalent to the ordinary least squares estimator:
    $$\hat{\beta} = (X'X)^{-1}X'y$$

* **Maximum Likelihood Estimator of Variance ($\tilde{\sigma}^2$):**
  $$\tilde{\sigma}^2 = \frac{(y - X\hat{\beta})'(y - X\hat{\beta})}{n}$$

* **Core Analytical Questions:**
  * What is the overall adequacy of the model?
  * Do specific regressors seem important within the network?

* **Formal Test Distribution Requirements:**
  * Random errors must be independent and follow a normal distribution.
  * Expected value of mean error condition: $E(\epsilon_i) = 0$
  * Variance of error condition: $\text{Var}(\epsilon_i) = \sigma^2$

#### **Test for Significance of Regression**

* **Core Objective Question:**
  * What is the significance of the regression test?

* **Operational Target Purpose:**
  * Determine if there is a linear relationship between the response variable $y$ and any regressor variable $x_1, x_2, \dots, x_k$.

#### **Hypothesis of Significance Testing**

* **Hypothesis Framework Configuration:**
  * **Null Hypothesis ($H_0$):** $\beta_1 = \beta_2 = \dots = \beta_k = 0$
  * **Alternative Hypothesis ($H_a$):** $\beta_j \neq 0$ for at least one $j$

* **Implications of Rejecting the Null Hypothesis:**
  * Rejection of the null hypothesis implies that at least one of the regressor variables $x_1, x_2, \dots, x_k$ contributes significantly to the model.
  * Alternatively, it states that at least one regressor variable is linearly dependent with the response variable.

* **Distribution Properties Under Rejection Conditions:**
  * If the null hypothesis is rejected or at least one $\beta_j \neq 0$, then $F_0$ follows a noncentral F-distribution.
  * Associated dimensions feature $k$ and $n - p$ (where $p = k + 1$) degrees of freedom ($\text{dof}$).

* **Noncentrality Parameter Calculation Matrix Formula ($\lambda$):**
  $$\lambda = \frac{\beta^{*'} X_c' X_c \beta^*}{\sigma^2}$$


#### **Centered Model Matrices and Analysis of Variance Test Statistics**

* **Parameter and Centered Matrix Components:**
  * Slope Parameter Vector: $\beta^* = (\beta_1, \beta_2, \dots, \beta_k)'$
  * $X_c$: Centered Model Matrix

* **Centered Model Matrix Structural Layout ($X_c$):**
  $$X_c = \begin{bmatrix}
  x_{11}-\bar{x}_1 & x_{12}-\bar{x}_2 & \dots & x_{1k}-\bar{x}_k \\
  x_{21}-\bar{x}_1 & x_{22}-\bar{x}_2 & \dots & x_{2k}-\bar{x}_k \\
  x_{31}-\bar{x}_1 & x_{32}-\bar{x}_2 & \dots & x_{3k}-\bar{x}_k \\
  \vdots & \vdots & & \vdots \\
  x_{n1}-\bar{x}_1 & x_{n2}-\bar{x}_2 & \dots & x_{nk}-\bar{x}_k
  \end{bmatrix}$$

#### **Sum of Squares Partitioning Equations**

* **Total Sum of Squares Formulation:**
  $$SST = SSR + SS_{\text{res}}$$
  * **$SST$**: Total sum of squares $\rightarrow \text{dof} = n - 1$
  * **$SSR$**: Regressor sum of squares $\rightarrow \text{dof} = k$
  * **$SS_{\text{res}}$**: Residual sum of squares $\rightarrow \text{dof} = n - (k + 1)$

#### **F-Test Ratio Calculation ($F_0$)**

* **Variance Ratio Test Statistic Matrix Equation:**
  $$F_0 = \frac{\frac{SSR}{k}}{\frac{SS_{\text{res}}}{n - (k + 1)}} = \frac{MSR}{MS_{\text{res}}}$$
  * This computed ratio tracks along an $F_{k, n-p}$ distribution pattern.

#### **Expected Values of Mean Squares**

* **Expected Value of Residual Mean Square:**
  $$E(MS_{\text{res}}) = \sigma^2$$

* **Expected Value of Regression Mean Square:**
  $$E(MSR) = \sigma^2 + \frac{\beta^{*'} X_c' X_c \beta^*}{k\sigma^2}$$


#### **F-Test Rejection Rules and ANOVA Summary Tables**

* **Rejection Decision Rule for $H_0$:**
  $$F_0 > F_{\alpha, k, n-(k+1)}$$
  * If this condition is true $\Rightarrow$ **Reject the Null Hypothesis $H_0$**.

#### **Computational Sum of Squares Matrix Equations**

* **Regression Sum of Squares ($SSR$):**
  $$SSR = \hat{\beta}'X'y - \frac{\left(\sum_{i=1}^{n} y_i\right)^2}{n} \quad \text{--- (1)}$$

* **Residual Sum of Squares ($SS_{\text{res}}$):**
  $$SS_{\text{res}} = SST - SSR \quad \text{--- (2)}$$

* **Total Sum of Squares ($SST$):**
  $$SST = y'y - \frac{\left(\sum_{i=1}^{n} y_i\right)^2}{n} \quad \text{--- (3)}$$
  *(Note: The notebook handwritten entry labels this third line as $SSR$, which represents a correction to standard total sum of squares notation $SST = y'y - \frac{(\sum y)^2}{n}$)*

| :--- | :--- | :--- | :--- | :--- |

#### **$R^2$ and Adjusted $R^2$: Checking Model Adequacy**

* **Behavior of Standard $R^2$:**
  * Generally, $R^2$ always increases when a regressor is added to a model.
  * This happens regardless of the actual structural value of the contribution of those variables.
  * If $R^2 \rightarrow 1.00$, it is typically considered a good model fit.

* **The Core Problem with $R^2$ and the Introduction of $R^2_{\text{Adj}}$:**
  * Because $R^2$ can be artificially inflated, the Adjusted $R^2$ ($R^2_{\text{Adj}}$) metric is introduced.
  * **Adjusted $R^2$ Matrix Equation:**
    $$R^2_{\text{Adj}} = 1 - \frac{\frac{SS_{\text{res}}}{n - p}}{\frac{SST}{n - 1}}$$

* **Operational Advantage of Adjusted $R^2$:**
  * $R^2_{\text{Adj}}$ will only increase upon adding a variable to the model if the addition of that variable reduces the overall residual mean square error.

* **Structural Component Breakdowns:**
  * **Residual Mean Square:**
    $$\frac{SS_{\text{res}}}{n - p} = MS_{\text{res}}$$
  * **Constant Base Variance:**
    $$\frac{SST}{n - 1} = \text{Constant term, regardless of how many variables are in the model.}$$

* **Conclusion:**
  * This penalty mechanism is how $R^2_{\text{Adj}}$ removes the artificial inflation problem inherent in standard $R^2$.

#### **Limitations of $R^2_{\text{Adj}}$ and Testing Individual Regression Coefficients**

* **Question:** So can we use $R^2_{\text{Adj}}$ for checking model adequacy?
* **Analysis of $R^2_{\text{Adj}}$ Limits:**
  * Sometimes $R^2_{\text{Adj}}$ penalizes us for adding a term that is not useful.
  * Additional terms can possibly cause overfitting in the model.
  * Therefore, we need to follow additional methods to check model adequacy.

#### **(A) Test on Individual Regression Coefficients**

* **Question:** Why do we require this test?
* **Core Rationale:**
  * Adding a variable to a regression model always increases the sum of squares for regression ($SSR$) and decreases the residual sum of squares ($SS_{\text{res}}$).

* **Impact of Adding a Variable Summary:**
  * Adding variable $\rightarrow$ Increased $SSR$
  * Adding variable $\rightarrow$ Decreased $SS_{\text{res}}$

#### **Variance Inflation and Individual Regressor Coefficient Tests**

* **Impact on Fitted Values:**
  * Adding unnecessary variables can increase the variance of the fitted value $\hat{y}$.
  * Therefore, we must be careful to include only regressors that are of real value in explaining the response.

* **Hypothesis Testing for Individual Regressor Coefficients ($\beta_j$):**
  * **Null Hypothesis ($H_0$):** $\beta_j = 0$
  * **Alternative Hypothesis ($H_a$):** $\beta_j \neq 0$

* **Implication of the Test Decision:**
  * If $H_0$ is not rejected, it indicates that the regressor $x_j$ can be deleted from the model.

* **T-Statistic Test to Check Contribution of the $j$-th Regressor Into Models:**
  $$t_0 = \frac{\hat{\beta}_j}{\sqrt{\hat{\sigma}^2 c_{jj}}} = \frac{\hat{\beta}_j}{\text{SE}(\hat{\beta}_j)}$$


#### **Covariance Diagonals, Rejection Regions, and Calculated Examples**

* **Structural Component Definitions:**
  * **$c_{jj}$**: The diagonal element of the inverse matrix $(X'X)^{-1}$ corresponding to the estimator $\hat{\beta}_j$.
  * **$\text{SE}(\hat{\beta}_j)$**: Represents the standard error of the estimator $\hat{\beta}_j$.
  * **$\hat{\sigma}^2$**: Error variance estimator computed via residual mean square:
    $$\hat{\sigma}^2 = MS_{\text{res}} = \frac{SS_{\text{res}}}{n - (k + 1)}$$

* **T-Test Rejection Criterion for $H_0: \beta_j = 0$:**
  * Reject the null hypothesis if the absolute value of the computed test statistic satisfies:
    $$|t_0| > t_{\alpha/2, n - k - 1}$$

#### **(B) Extra Sum of Squares Method**

* **Core Analytical Objective:**
  * Determine the individual or joint contribution of a regressor or a group of regressors to the overall regression sum of squares.

* **Operational Mechanism:**
  * This method is used to investigate the specific contribution of a subset of regression variables.
  * Given regression variables: $x_1, x_2, x_3, \dots, x_k$
  * Where a specific group (e.g., $x_2, x_3$) is treated as a isolated **subset**.
  * This method formally demonstrates how to calculate the joint contribution of regressors $x_2$ and $x_3$ into the models.

* **Step I: Calculate Marginal Regression Sum of Squares**
  * Compute the regressor sum of squares due to a subset of regressors.
  * **Notation Expression:** $SSR(\beta_2 | \beta_1)$
  * This represents the extra sum of squares due to $\beta_2$ given that $\beta_1$ is already included inside the model structure.

* **Step II: Calculate F-Statistic**
  * Compute the corresponding partial F-test statistic to evaluate significance.

#### **Partial F-Test Ratios and Noncentrality Matrix Constraints**

* **Partial F-Statistic Tensor Equation ($F_0$):**
  $$F_0 = \frac{\frac{SSR(\beta_2 | \beta_1)}{\text{dof}}}{MS_{\text{res}}}$$
  * **$MS_{\text{res}}$**: Residual mean square computed due to the complete model layout.
  * **$\text{dof}$**: Degrees of freedom matching the isolated subset of regressors.

* **Degrees of Freedom Assignment Behavior:**
  * If the subset includes exactly $1$ regressor variable $\Rightarrow \text{dof} = 1$.
  * If the subset includes exactly $3$ regressor variables $\Rightarrow \text{dof} = 3$.

* **Distribution Properties Under Rejection:**
  * If the null hypothesis is rejected, then $F_0$ follows a noncentral F-distribution.

* **Step III: Noncentrality Parameter Calculation Matrix Formula ($\lambda$):**
  * If the null hypothesis $H_0: \beta_2 = 0$ is rejected based on the critical bounding criterion:
    $$F_0 > F_{\alpha, r, n-p}$$
  * Then the noncentral parameter is calculated via the matrix product sequence:
    $$\lambda = \frac{1}{\sigma^2} \beta_2' X_2' \left[ I - X_1 (X_1' X_1)^{-1} X_1' \right] X_2 \beta_2$$

* **Maximum Statistical Power Conditions:**
  * **Question:** When does the maximum power of this test occur?
  * **Answer:** Maximum test power is achieved when the partitioned model sub-matrices $X_1$ and $X_2$ form an orthogonal matrix pair relative to each other.
  * **Orthogonality Metric Condition:**
    $$X_2' X_1 = 0$$

  * **Definitive Conclusion:** We are $95\%$ statistically confident that the true population mean delivery time spans within the range of $17.66 \text{ minutes}$ to $20.78 \text{ minutes}$, and the fitted point output value also resides safely inside these calculated confidence bounds.

#### **Simultaneous Confidence Intervals on Regression Coefficients**

* **Core Operational Definition:**
  * Confidence Coefficient $\Rightarrow (1 - \alpha)$

* **Context and Core Rationale:**
  * Previously, individual confidence intervals and prediction intervals were calculated one-at-a-time.
  * The individual confidence coefficient $(1 - \alpha)$ indicates the proportion of correct statements when repeated random samples are selected and appropriate interval estimates are constructed for each separate sample.
  * **The Problem:** A new challenge arises when we construct several confidence or prediction intervals using the same underlying sample dataset.

* **Defining Joint Intervals:**
  * A set of separate confidence intervals that are all true simultaneously with a shared joint probability of $(1 - \alpha)$ are formally called **simultaneous** or **joint confidence intervals**.

* **Simultaneous Interval Estimation Matrix Formula:**
  * The simultaneous interval estimate of the parameters in a linear regression model is constructed via:
    $$\hat{\beta}_j \pm \Delta \, \text{SE}(\hat{\beta}_j)$$
  * **Standard Error Component Matrix Calculation:**
    $$\text{SE}(\hat{\beta}_j) = \sqrt{\hat{\sigma}^2 c_{jj}}$$
  * **Unbiased Residual Error Standard Deviation Equation:**
    $$\hat{\sigma} = \sqrt{\frac{MS_{\text{res}}}{n - p}}$$

#### **Bonferroni Method and Scheffé S-Method for Multi-Interval Testing**

* **The Bonferroni Method for Calculating Critical Multipliers ($\Delta$):**
  * Baseline Confidence Coefficient $\Rightarrow (1 - \alpha)$
  * Adjusted Bonferroni Multiplier Confidence Base $\Rightarrow 1 - \left(\frac{\alpha}{p}\right)$

* **Bonferroni Delta ($\Delta$) Multiplier Equation:**
  $$\Delta = t_{\alpha/2p, \, n-p}$$

* **Definitive Bonferroni Simultaneous Confidence / Prediction Interval Matrix Formula:**
  $$\hat{\beta}_j \pm t_{\alpha/2p, \, n-p} \times \text{SE}(\hat{\beta}_j)$$

#### **Alternative Joint Interval Adjustments**

* **Scheffé S-Method Multiplier Equation:**
  $$\Delta = \left( 2F_{\alpha, \, p, \, n-p} \right)^{1/2}$$

* **Maximum Modulus t Procedure Multiplier Equation:**
  $$\Delta = u_{\alpha, \, p, \, n-p}$$

* **Step 1: Determine Maximum Leverage Bounds ($h_{\text{max}}$):**
  * Find $h_{\text{max}}$, which represents the maximum numerical value found along the diagonal elements of the training hat matrix ($H$).

#### WHY DO REGRESSION COEFFICIENTS HAVE THE WRONG SIGN?

When using multiple regression, occasionally we find an apparent contradiction of intuition or theory when one or more of the regression coefficients seem to have the wrong sign. For example, the problem situation may imply that a particular too much their direction. Suppose QQ regression coefficient should be positive, while the actual estimate of the parameter is negative. This “wrong”-sign problem can be disconcerting, as it is usually difficult to explain a negative estimate (say) of a parameter to the model user when that user believes that the coefficient should be positive. 

1. The range of some of the regressors is too small.
2. Important regressors have not been included in the model.
3. Multicollinearity is present.
4. Computational errors have been made.

#### Impact of the Range of $X$ on Regression Coefficients

It is easy to see how the **range of the $X$'s** can affect the **sign of the regression coefficients**. 

Consider the simple linear regression model:
$$Y = \beta_0 + \beta_1 X + \epsilon$$

The variance of the regression coefficient $\widehat{\beta}_1$ is given by:
$$\text{Var}(\widehat{\beta}_1) = \frac{\sigma^2}{\sum_{i=1}^{n} (X_i - \bar{X})^2}$$


### Sampling Distribution and Variance of $\widehat{\beta}_1$

![Sampling Distribution and Variance](image-1.png)

Note that the variance of $\widehat{\beta}_1$ is **inversely proportional to the "spread" of the regressor**. Therefore, if the levels of $X$ are all close together, the variance of $\widehat{\beta}_1$ will be relatively large. 

In some cases, the variance of $\widehat{\beta}_1$ could be so large that a negative estimate (for example) of a regression coefficient that is really positive results. 

The situation is illustrated in **Figure 3.14** which plots the sampling distribution of $\widehat{\beta}_1$. Examining this figure, we see that the probability of obtaining a negative estimate of $\widehat{\beta}_1$ depends on:
1. How close the true regression coefficient ($\beta_1$) is to zero.
2. The variance of $\widehat{\beta}_1$, which is greatly influenced by the spread of the $X$'s.

### Design of Experiments and Omitted Variable Bias

In some situations, the analyst can control the levels of the regressors. Although it is possible in these cases to decrease the variance of the regression coefficients by increasing the range of the $X$'s, it may not be desirable to spread the levels of the regressors out too far:
* **Curvature/Nonlinearity:** If the $X$'s cover too large a range and the true response function is nonlinear, the analyst may have to develop a much more complex equation to adequately model the curvature in the system.
* **Region of Interest:** Many problems involve a region of $X$ space of specific interest to the experimenter, and spreading the regressors out beyond this region of interest may be impractical or impossible.

**The Trade-off:** In general, we trade off the precision of estimation, the likely complexity of the model, and the values of the regressors of practical interest when deciding how far to spread out the $X$'s.

#### Omitted Variable Bias (Sign Reversal)
Wrong signs can also occur when **important regressors have been left out of the model**. In these cases, the sign is not really wrong; the partial nature of the regression coefficients causes the sign reversal (Simpson's Paradox / Omitted Variable Bias).

![Variable Bias](image-2.png)

Suppose we fit a model involving only y and x1. The equation is
#### Total vs. Partial Regression Coefficients (Sign Reversal Example)

Consider the simple linear regression model containing only $x_1$:
$$\hat{y} = 1.835 + 0.463x_1$$

Where $\widehat{\beta}_1 = 0.463$ is a **"total" regression coefficient**. That is, it measures the total effect of $x_1$ ignoring the information content in $x_2$. 

Now consider the multiple regression model involving both $x_1$ and $x_2$:
$$\hat{y} = 1.036 - 1.222x_1 + 3.649x_2$$

Note that now $\widehat{\beta}_1 = -1.222$, and a **sign reversal** has occurred. 

The reason is that $\widehat{\beta}_1 = -1.222$ in the multiple regression model is a **partial regression coefficient**; it measures the effect of $x_1$ given that $x_2$ is also in the model.

**Interpretation from Figure 3.15:**
* If we ignore the $x_2$ values, the apparent relationship between $y$ and $x_1$ has a **positive slope**.
* However, if we consider the relationship between $y$ and $x_1$ for constant values of $x_2$, this relationship really has a **negative slope**. 

Thus, a wrong sign in a regression model may indicate that an important regressor has been omitted.

### Multicollinearity and Computational Error as Sources of Wrong Signs

If important regressors are missing, the analyst can identify these regressors and include them in the model, and then the wrong signs may disappear.

#### Multicollinearity
**Multicollinearity** can cause wrong signs for regression coefficients. In effect, severe multicollinearity inflates the variances of the regression coefficients, and this increases the probability that one or more regression coefficients will have the wrong sign. Methods for diagnosing and dealing with multicollinearity are summarized in **Chapter 9**.

#### Computational Error
Computational error is also a source of wrong signs in regression models:
* Different computer programs handle **round-off** or **truncation** problems in different ways, and some programs are more effective than others in this regard. 
* Severe multicollinearity causes the $X'X$ matrix to be **ill-conditioned**, which is a significant source of computational numerical instability. 
* Computational error can cause not only sign reversal maximum-likelihood results but regression coefficients to differ by several orders of magnitude. 

The accuracy of the computer code should be investigated when wrong-sign problems are suspected.
