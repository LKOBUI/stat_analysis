In statsmodels, you can fit a Generalized Least Squares (GLS) model when you have a known covariance structure for the errors.

Generalized Least Squares (GLS):

1. Purpose: GLS is used when the errors are not independent or have non-constant variance (heteroscedasticity) with a known structure.
2. statsmodels Implementation: statsmodels.regression.linear_model.GLS allows fitting a GLS model.

Notes:

1. Known Covariance: GLS requires a known covariance structure for the errors.
2. Feasible GLS: ***If the covariance structure is unknown, Feasible GLS can be used with an estimated covariance***.

```python
dataset = sm.datasets.longley.load()
df = dataset.data
print(f" df.columns name {df.columns}")
X = df.drop(['TOTEMP'], axis=1)
print(f"Shape {X.shape}")
y = df['TOTEMP']
X = sm.add_constant(X)
print(f"Shape {X.shape}")
# mod = smf.ols(formula='Lottery ~ Literacy + Wealth + Region', data=df)
```

#### Prepare models for rho calculations

```python
# Get residuals
resid_1 = sm.OLS(y, X).fit().resid

print(f"Shape of resid_1 {resid_1.shape}")

# Prepare dtase accordin AR(1) models to Calculate rho
y_resid = np.asarray(resid_1)[1:]
X_resid = sm.add_constant(np.asarray(resid_1)[:-1])
print(f"Shape of y_resid {y_resid.shape}, X_resid: {X_resid.shape}")
```

#### Apply AR1 Process and calculate rho

```python

# εi = resid_fit 
resid_fit = sm.OLS(y_resid,X_resid).fit()
df_AR1 = pd.DataFrame({'coff':resid_fit.params,
                       't scor':resid_fit.tvalues,
                       'p val':resid_fit.pvalues},
                      index=['β0','ρ'])
rho = resid_fit.params[1]
tableDispFormatt('AR1 Process Models Parameters')
print(df_AR1)
print(f"ρ value of AR1 Models {rho}")

if (resid_fit.pvalues[1]) <= 0.05:
    print("Nul Hypothesis Rejected")
else:
    print("Fail to reject Null hypothesis!!!")
    print("we do not have strong evidence that the errors follow an AR(1) process")
    print("So we need to contine")
```

```python
# Calculate Order

order = toeplitz(range(len(resid_1)))
order,order.shape
```

#### Get autocorrelation structure

```python
sigma = rho**order
gls_model = sm.GLS(y, X, sigma=sigma)
gls_results = gls_model.fit()
print(gls_results.summary())
```

#### Use GLSAR with one lag, to get to a similar result:

```python
glsar_model = sm.GLSAR(y, X, rho = 1)
glsar_results = glsar_model.iterative_fit( maxiter=10)
print(glsar_results.summary())
```
