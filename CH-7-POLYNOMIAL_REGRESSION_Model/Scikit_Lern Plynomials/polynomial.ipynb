Polynomial just provide transformations on Data set. It not provides any kind of regressions.
It's taken mainaly 3 arguments
PolynomialFeatures(degree=2, # just provide degree of regressors
                    interaction_only=False, 
                    include_bias=True, 
                    order='C' # Not includes
                    )
1. include_bias = True it will add const ter terms in data set just like sm.add_constant()
2. interaction_only = True
   let regressors = [u v]
   and degree = 2
   if interaction_only = False; 
      include_bias = True
   Const + u + v + u2 + uv + v2 (with degree = 2)
   Const + u + v + u2 + uv + v2 + u3 + u2v + uv2 + v3 (with degree = 3)

   if interaction_only = True  (ingnore higher degree terms); 
      include_bias     = False ()
   u + v + uv  (with degree = 2)
   u + v + uv  (with degree = 3)
   u + v + uv  (with degree = 4)  [Regressions eqtions are all same, no effect]

   

```python
X = np.arange(6).reshape(3, 2)
X
```

```python
obj = PolynomialFeatures(degree = 3, interaction_only = False, include_bias=True)
poly = obj.fit(X)
res = poly.transform(X)
res
```

```python
obj = PolynomialFeatures(degree = 3, interaction_only = True, include_bias=True)
poly = obj.fit(X)
res = poly.transform(X)
res
```
