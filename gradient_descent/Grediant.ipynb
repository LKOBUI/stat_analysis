### Batch Gradient Descent

Partial derivatives of the cost function



Gradient vector of the cost function






```python
# Generate some sample data
np.random.seed(0)
m = 100
X = np.random.rand(m, 1)
y = 3 * X + 2 + np.random.randn(m, 1)
# Add a column of ones to X for the bias term
X = np.hstack((np.ones((m, 1)), X))
```

```python
BETAS_PATH = [] #further use
LR = 0.01
NEPOCH = 500
def BatchGrediantDesent(x_var,y_var,lerning_rate,n_epochs):
    fX = sm.add_constant(x_var)
    # Initialize the parameters
    # randomly initialized model parameters
    beta = np.random.randn(fX.shape[1], 1)  
    for epoch in range(n_epochs):
        gradients = 2 / fX.shape[0] * fX.T @ (fX @ beta - y_var)
        beta = beta - lerning_rate * gradients
        BETAS_PATH.append(beta)
    return beta
    
BatchGrediantDesent(
    x_var = X,
    y_var = y,
    lerning_rate = LR,
    n_epochs = NEPOCH 
)
```

### Stochastic Gradient Descent

Stochastic Gradient Descent just picks a random instance in the training set at every step and computes the gradients based only on that single instance.

***Stochastic Gradient Descen***t has a ***better chance*** of ***finding the global minimum*** than Batch Gradient Descent does.

##### learning schedule:
The function that determines the learning rate at each iteration is called the learning schedule. If the learning rate is reduced too quickly, you may get stuck in a local minimum, or even end up frozen halfway to the minimum. If the learning rate is reduced too slowly, you may jump around the minimum for a long time and end up with a suboptimal solution if you halt training too early.

### Mini-batch Gradient Descent

What is Mini-batch Gradient Descent?

Mini-batch Gradient Descent is an optimization algorithm that uses a small batch of training data to compute the gradient of the loss function with respect to the model's parameters. The model's parameters are then updated based on this gradient.
