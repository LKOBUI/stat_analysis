### matplotlib.pyplot.contour





```python
# Example: Figure 3.1 from book Introductions to Linear Regression Analysis, Pg 68
# Equation 3.1 E(y) = 50 + 10x1 + 7x2

# Step 1: Generate Data
samples = 20
X1 = np.linspace(0,10,samples) + np.random.randn(samples)
Y1 = X1*7
# Step 2: Mesh grid
X1,Y1 = np.meshgrid(X1,Y1)

Z1 = 1 + 10*X1 + Y1
Z1.shape, X1.shape,Y1.shape
```

```python
fig, ax = plt.subplots()
CS = ax.contour(X1, Y1, Z1)
ax.clabel(CS, fontsize=10)
ax.set_title('Simplest default with labels')
```
