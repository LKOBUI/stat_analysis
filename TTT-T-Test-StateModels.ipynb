```python
def getRegressionEqn(pdSeries):
    eqnStr = str()
    idx = (pdSeries.index).to_list()
    val = (pdSeries.values).tolist()
    for i in np.arange(0,len(idx),1):
        if val[i] >= 0.0:
            if i > 0:
                eqnStr += '+' + ' ' +str(val[i])+' '+ idx[i] + ' '
            else:
                eqnStr += str(val[i])+' '+ idx[i] + ' '
        else:
            if i > 0:
                eqnStr +=  str(val[i])+' '+ idx[i] + ' '
            else:
                eqnStr +=  str(val[i])+' '+ idx[i] + ' '
    print(eqnStr)
    return eqnStr
```
