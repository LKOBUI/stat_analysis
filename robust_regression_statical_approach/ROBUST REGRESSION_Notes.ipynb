### Robust Regression

Robust regression refers to a set of regression techniques designed to be less sensitive to outliers and violations of model assumptions (such as non-normal errors or heteroscedasticity) than ordinary least squares (OLS) regression.

#### Why Use Robust Regression?

•	OLS regression can be heavily influenced by outliers or extreme values, which can distort parameter estimates and predictions.
•	Robust regression methods reduce the influence of outliers, providing more reliable estimates when data are messy or assumptions are violated.

#### Key Concepts in Robust Regression

1. Down weighting Outliers

•	Robust methods do not treat all data points equally.
•	They assign less weight to points that are far from the fitted model, reducing the influence of outliers.

2. Alternative Loss Functions

	- Ordinary Least Squares (OLS) minimizes the sum of squared residuals: Σr_i^2, which penalizes large errors heavily. Squaring each residual means that **larger errors contribute disproportionately to the total loss**:
	 - A residual of 2 contributes 2^2 = 4
	 - A residual of 10 contributes 10^2 = 100
    
     So, OLS is very sensitive to outliers—a single large error can dominate the loss function and skew the model.

	- Note: To reduce this sensitivity, robust methods like LAD (Least Absolute Deviations) or Huber loss use different loss functions that grow more slowly for large residuals.

	- Robust methods use loss functions that grow more slowly for large residuals, reducing the impact of extreme errors.

#### Common Robust Regression Procedures

1. M-estimators
	- Generalize maximum likelihood estimation.
	- Minimize a function of the residuals, often less sensitive to large errors.
	- Huber loss is a popular choice:
    - Quadratic for small residuals, linear for large residuals.
    - Tukey’s biweight and Hampel’s loss are other examples.
2. Least Absolute Deviations (LAD) Regression
	- Minimizes the sum of absolute residuals Σ|r_i |, not squared residuals.
	- More robust to outliers than OLS.
3. RANSAC (Random Sample Consensus)
-	Iteratively fits the model to random subsets of the data, identifying inliers and outliers.
-	Only inliers are used to estimate the final model.
4. MM-estimators
-	Combine high breakdown point (resistance to outliers) and high efficiency (good performance when data are clean).

General Robust Regression Procedure

1.	Choose a robust method (e.g., Huber, LAD, RANSAC)
2.	Fit the model using the chosen method
3.	Iteratively reweight data points based on their residuals (for M-estimators)
4.	Estimate coefficients that minimize the robust loss function
5.	Assess model fit and check residuals for remaining outliers

In Summary

•	Robust regression methods protect against outliers and model assumption violations.
•	They use alternative loss functions or weighting schemes to reduce the influence of problematic data points.
•	Common methods include M-estimators (Huber, Tukey), LAD, and RANSAC.

### L1 Regularization (Lasso Regression):

L1 regularization is used to prevent overfitting by adding a penalty to the loss function based on the absolute values of the model coefficients.

### Key Points about L1 Regularization
 - Also known as: Lasso (Least Absolute Shrinkage and Selection Operator)
 - Penalty term: The sum of the absolute values of the coefficients Σ|β_j|
 - Effect: Encourages sparsity—many coefficients are shrunk exactly to zero, effectively performing variable selection



### How Does L1 Regularization Work?

•	When λ = 0: No regularization, standard least squares.
•	As λ increases: More coefficients are shrunk toward zero; some may become exactly zero.
•	Lasso regression can reduce coefficient values to zero. This leads to feature selection—only the most important predictors remain in the model.

### Applications

•	High-dimensional data: When you have many predictors, L1 regularization helps select a subset.
•	Sparse models: Useful when you believe only a few predictors are truly important.



### Summary

•	L1 regularization adds a penalty for large coefficients, encouraging sparsity.
•	It is the basis for Lasso regression, which can perform variable selection.
•	It is widely used in machine learning and statistics to improve model generalization and interpretability.

### How to Choose the Correct λ (lambda) in L1 Regularization?

1.	Step 1: Define a grid of possible λ values (e.g., 0.001, 0.01, 0.1, 1, 10)
2.	Step 2: For each λ, fit the Lasso model on the training data
3.	Step 3: Evaluate the model’s performance (e.g., mean squared error) on validation data (or using k-fold cross-validation)
4.	Step 4: Choose the λ that gives the best validation performance (lowest error)

### L2 regularization, or ridge regression:

**L2 regularization, or ridge regression**, is a regularization method that ***penalizes high-value coefficients*** in a machine learning model similarly to L1 regularization — though with a different penalty term. L2 regularization adds the squared sum of coefficients (or the “squared magnitude” of the coefficient) as the penalty term to the model’s **SSE loss function**.



L2 regularization can reduce coefficient values toward zero but never exactly to zero. This also means that L2 regularization cannot perform feature selection. 

- L2 regularization can effectively handle multicollinearity in a data set. 
- if lambda is very large then it will add too much weight and lead to underfitting.
