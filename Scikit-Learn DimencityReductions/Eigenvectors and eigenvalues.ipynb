### Qus: what should be rank of the strongly correlated collinear matrix ?

-	Matrix rank = number of linearly independent columns (or rows)
-	Collinear columns = columns that are scalar multiples or linear combinations of each other
-	Strong correlation (e.g., correlation ≈ ±1) often implies near collinearity, especially in regression

### So What Should the Rank Be?

-	If perfectly collinear → rank = 1
-	If strongly correlated but not exact → rank = 2 (or slightly less than full)
-	If uncorrelated → rank = full (e.g., 3 for a 3-column matrix)



### Qus: What is Eigenvector?

Eigenvector are those vectors whose do not change their rotations about to origin when they multiply to any matrix vectors. They can Strach and shrink on vector span.

In simpler terms, the eigenvector retains its original direction after being transformed by the matrix, even though its length might be scaled. 

When a vector changes direction after multiplication with a matrix, it is concluded that the vector is not an eigenvector of that matrix.
