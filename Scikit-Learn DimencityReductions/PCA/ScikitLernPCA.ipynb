```python
# get 2-D  matrix random number between 1 to 10 total 5 component
def get_random_matrix(rows, cols):
    return np.random.randint(1, 20, size=(rows, cols))
X = get_random_matrix(5, 2)
#scatter plot
def scatter_plot(X, y=None):
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', edgecolor='k')
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.title('Scatter Plot of Features')
    plt.show()
scatter_plot(X)
```
