#### Cross Validation in Machine Learning (Prevent Overfitting)

- Cross-validation is a technique used to check **how well a machine learning model performs**on unseen data. 
- It s**plits the data into several parts**, trains the model on some parts and tests it on the remaining 
  part repeating this process multiple times. 

The **main purpose of cross validation is to prevent overfitting**.

Evaluate a score by **cross-validation**.

KFold is a cross-validation technique used in machine learning to evaluate the performance of a model.

Why Use KFold?

1. Evaluate model performance
2. Reduce overfitting
3. Hyperparameter tuning

How KFold Works ?

1. Split data: The dataset is split into k subsets or "folds".
2. Train and test: The model is trained on k-1 folds and tested on the remaining fold.
3. Repeat: Steps 1 and 2 are repeated k times, with each fold being used as the test set once.

Types of KFold

1. KFold
2. Stratified KFold

### KFold:

```python
X = np.random.randint(10,20,10).reshape(5,2)
y = np.array([1,2,3,4,5])
```

```python

kf = KFold(n_splits = 2)
kf.get_n_splits(X)
for i, (train_index, test_index) in enumerate(kf.split(X)):
    print(f"Fold {i}:")
    print(f"  Train: index={train_index}")
    print(f"  Test:  index={test_index}")
```

### StratifiedKFold

StratifiedKFold, on the other hand, preserves the class balance in each fold. This ensures that the ***proportion of samples from each class is maintained in each fold***, making it ***suitable for classification*** problems with imbalanced classes.

Understand difference b/w KFold and StratifiedKFold:

Example
Suppose you have a dataset with 100 samples, 80 of which belong to class A and 20 of which belong to class B. If you use KFold with k=5, the folds might look like this:

Fold 1: 16 samples from class A, 4 samples from class B
Fold 2: 20 samples from class A, 0 samples from class B
...

With StratifiedKFold, the folds would be split to ***preserve the class balance***:

Fold 1: 16 samples from class A, 4 samples from class B
Fold 2: 16 samples from class A, 4 samples from class B

When to Use Each
1. Use KFold:
    - When the classes are balanced or the problem is regression.
    - When the dataset is large and the class balance is not a concern.
2. Use StratifiedKFold:
    - When the classes are imbalanced and you want to ***preserve the class balance in each fold***.
    - When you're working on a classification problem and want to ensure that the model is 
      evaluated fairly.
