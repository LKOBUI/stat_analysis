### Precision and Recall

#### Precision (measures the accuracy of positive predictions):
Definition:

Out of all the items the model predicted as positive, how many were actually positive? 

Formula:
Precision = True Positives / (True Positives + False Positives) 
Example:
If a spam filter correctly identifies 90 out of 100 emails it flagged as spam, the precision is 90%. 

#### Importance:
High precision is important when the cost of a false positive is high, such as in medical diagnosis or fraud detection. 

### Recall (also known as Sensitivity):

Definition:

Out of all the actual positive instances, how many did the model correctly identify? 

Formula:

Recall = True Positives / (True Positives + False Negatives) 

Example:

If a cancer screening test correctly identifies 80% of all actual cancer cases, the recall is 80%. 

#### Importance:

High recall is important when the cost of a false negative is high, such as in disease detection or security systems.

### F1 Score: [harmonic mean of precision and recall]

What is F1 Score?

The F1 score is the harmonic mean of precision and recall. It provides a balanced measure of both precision and recall.

Why Use F1 Score?

**1. Balances precision and recall**: The F1 score provides a balanced measure of both precision and recall.

**2. Handles class imbalance:** The F1 score is useful for evaluating models on datasets with class imbalance.

**3. Provides a single metric:** The F1 score provides a single metric to evaluate the performance of a model.

When to Use F1 Score

**1. Classification problems**: The F1 score is commonly used to evaluate the performance of classification models.

**2. Imbalanced datasets**: The F1 score is particularly useful for evaluating models on datasets with class imbalance.



## **Trad-OFF**

**Precision**

Precision is the ratio of true positives to the sum of true positives and false positives. It measures the accuracy of the model's positive predictions.

**Recall**

Recall is the ratio of true positives to the sum of true positives and false negatives. It measures the model's ability to detect all instances of the positive class.

**Trade-off**

The precision/recall trade-off arises because increasing precision often decreases recall, and vice versa. This is because:

- Increasing precision requires reducing false positives, which may lead to more false negatives and decreased recall.
- Increasing recall requires reducing false negatives, which may lead to more false positives and decreased precision.

That's why tread-off come into picture.

#### Qus: How to select threshold ?

**Method 1: Use decision_function() API instade of classifier’s predict() method**

Scikit-Learn does not let you set the threshold directly, but it does give you access to the decision scores that it uses to make predictions. Instead of calling the classifier’s predict() method, you can call its decision_function() method, which returns a score for each instance, and then make predictions based on those scores using any threshold you want.


```python
from sklearn.datasets import fetch_openml
mnist = fetch_openml('mnist_784', as_frame=False)
X, y = mnist.data, mnist.target

# Get Traning & Test Data
X_train, X_test, y_train, y_test = X[:60000], X[60000:], y[:60000], y[60000:]

# Create models
models = SGDClassifier(random_state=42)
sgd_clf = models.fit(X_train, y_train)
```

```python
some_digit = X[0]
some_digit.shape
#some_digit
```

```python
# Method 1: Use decision_function() API instade of classifier’s predict() method
y_scores = sgd_clf.decision_function([some_digit])
sgd_clf.predict([some_digit])
```

```python
threshold = 0
y_some_digit_pred = (y_scores > threshold)
y_some_digit_pred
```

```python
threshold = 3000
y_some_digit_pred = (y_scores > threshold)
y_some_digit_pred
```

### Now how do you decide which threshold to use?

#### **Method 2: Use cross_val_predict() function**

 - Note: cross_val_predict() API differenty from cross_val_score().

```python
y_train_5 = (y_train == '5')  # True for all 5s, False for all other digits
y_test_5 = (y_test == '5')
y_scores = cross_val_predict(sgd_clf, X_train, y_train_5, cv=3,
                             method="decision_function")
```

```python
precisions, recalls, thresholds = precision_recall_curve(y_train_5, y_scores)
print(precisions.shape,recalls.shape, thresholds.shape)
```

#### Precision and recall versus the decision threshold

```python
plt.figure(figsize=(8, 4))  # extra code – it's not needed, just formatting
plt.plot(thresholds, precisions[:-1], "b--", label="Precision", linewidth=2)
plt.plot(thresholds, recalls[:-1], "g-", label="Recall", linewidth=2)
plt.vlines(threshold, 0, 1.0, "k", "dotted", label="threshold")

# extra code – this section just beautifies and saves Figure 3–5
idx = (thresholds >= threshold).argmax()  # first index ≥ threshold
plt.plot(thresholds[idx], precisions[idx], "bd")
plt.plot(thresholds[idx], recalls[idx], "gd")
plt.axis([-50000, 50000, 0, 1])
plt.grid()
plt.xlabel("Threshold")
plt.legend(loc="center right")
#save_fig("precision_recall_vs_threshold_plot")

plt.show()
```
