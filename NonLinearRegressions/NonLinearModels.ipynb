### Qus: What is problems with the regression model (Y = Xβ + ε) when response variable Y is binary ?

1. OLS Assumes Continuous Output
-	OLS regression is designed for continuous response variables.
-	It predicts values anywhere on the real number line, not just 0 or 1.
2. Predicted Values Can Be Outside [0, 1]
-	For binary outcomes, the true values are only 0 or 1. 
-	OLS can predict values less than 0 or greater than 1, which do not make sense as probabilities.
3. Violation of Model Assumptions
-	Homoscedasticity:
OLS assumes constant variance of errors. For binary data, the variance depends on the mean, specifically p(1 - p), so this assumption is violated.
