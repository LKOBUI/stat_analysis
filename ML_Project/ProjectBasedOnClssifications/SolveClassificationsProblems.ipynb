# Popular Classification Algorithms in ML




How to Choose the Right Algorithm:

- Small dataset, linear boundary → Logistic Regression, SVM
- Text classification → Naïve Bayes, Logistic Regression
- Tabular data with mixed features → Random Forest, Gradient Boosting
- Image/audio/NLP → Neural Networks (CNNs, RNNs, Transformers)
- Quick baseline → Decision Tree or Logistic Regression


Core Validation Strategies
- ### Train/Test Split (Holdout Method)
  - Split dataset into training and testing sets (e.g., 80/20).
  - Train on one, evaluate on the other.
  - Simple but can be sensitive to how the split is made.
- ### Cross-Validation (CV)
  - **k-Fold CV**: Split data into k folds (e.g., 5 or 10). Train on k-1 folds, test on the remaining fold, repeat, and average results.
  - **Stratified k-Fold:** Ensures class proportions are preserved in each fold (important for imbalanced datasets).
  - More robust than a single train/test split.
- ### Leave-One-Out CV (LOOCV)
  - **Extreme case of CV**: each sample is tested once while the rest are used for training.
  - Very thorough but computationally expensive.


### Key Evaluation Metrics

#### Classification Model Evaluation Metrics:

- Accuracy → Percentage of correct predictions.
  - Use when classes are balanced.
- Precision → Of predicted positives, how many are actually correct.
  - Important when false positives are costly (e.g., spam detection).
- Recall (Sensitivity) → Of actual positives, how many were correctly identified.
  - Important when false negatives are costly (e.g., disease detection).
- F1-Score → Harmonic mean of precision and recall.
  - Good balance when classes are imbalanced.
- ROC Curve & AUC → Shows trade-off between true positive rate and false positive rate.
  - Useful for binary classifiers.
- Confusion Matrix → Table showing counts of True Positives, False Positives, True Negatives, and False Negatives.
  - Helps visualize errors.

#### Qus: Explain statsmodels.base.model.LikelihoodModelResults how this work

The **statsmodels.base.model.LikelihoodModelResults** class is a base results container used in Statsmodels to hold the output of models **estimated via maximum likelihood**. You don’t use it directly—instead, **it’s inherited by more specific results classes like GLMResults, LogitResults, or OLSResults**.

#### What Is LikelihoodModelResults?

It’s a foundational class that stores:

- Estimated parameters
- Covariance matrices
- Fit statistics
- Optimization details

It’s returned when you fit a model like GLM, Logit, Probit, etc., and call .fit()

#### When you call .fit() on a likelihood-based model, Statsmodels:

- Runs an optimization routine (e.g., Newton-Raphson, BFGS)
- Stores convergence info in mle_retvals
- Computes standard errors from the Hessian or observed information matrix

This structure makes it easy to extend or customize models while keeping a consistent interface for results.

#### Qus what is  dispersion parameter ?

The dispersion parameter in a Generalized Linear Model (GLM) quantifies ***how much the observed data deviates from the model’s expected variance***. It acts as a scale factor for the variance of the response variable.


What Is "Scale" in GLMs?

- The scale parameter reflects the variability of the data not explained by the model.
- For some families (like Gaussian), it’s estimated from the residuals.
- For others (like Binomial or Poisson), the scale is fixed at 1 by default because the variance is fully determined by the mean.
