### softmax formula calculates the probability for each class in a multi-class classification problem



```python

def Softmax(z):
    exp_z = np.exp(z - np.max(z))  # stability trick
    return exp_z / np.sum(exp_z)
```

```python
path = os.path.join(os.getcwd(), 'Iris.csv')
df = pd.read_csv(path)
df.drop(['Id'], axis=1, inplace=True)
df.columns
```

```python
X = df.drop(['Species'], axis=1, inplace=False)
X = sm.add_constant(X)
y = df['Species'].map({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
```

```python
# One-hot encode target
K = len(np.unique(y))  # number of classes
Y = np.eye(K)[y]  # shape (150, 3)
```

```python
# Initial guess. Youi can take any value
beta_init = np.zeros((K, X.shape[1])).flatten()
```

```python
from scipy.special import softmax
# Define negative log-likelihood for multinomial logistic regression
def Nllm(beta_flat,n_class):
    beta = beta_flat.reshape(n_class, -1)  # shape (3, 5)
    logits = X @ beta.T  # shape (150, 3)
    probs = softmax(logits, axis=1)  # shape (150, 3)    
    # The log-likelihood formula for logistic regression
    # LL = sum(y * log(p) + (1 - y) * log(1 - p))
    log_likelihood = np.sum(Y * np.log(probs) + (1 - Y) * np.log(1 - probs))
    #print(f"Log Likelihood: {log_likelihood}")
    return -log_likelihood
```

### Optimizations Constrains

```python
# Optimize
from scipy.optimize import minimize
result = minimize(Nllm, beta_init,K, method='BFGS')
beta_hat = result.x.reshape(K, -1)
print(beta_hat)
```

#### Predictions and models accuracy check

```python
# Predict
logits = X @ beta_hat.T
probs = softmax(logits, axis=1)
y_pred = np.argmax(probs, axis=1)

# Accuracy
accuracy = np.mean(y_pred == y)
print(f"Multinomial GLM accuracy: {accuracy:.4f}")
```

#### Step by Step Implementations of IRLS

The genius of IRLS is that it turns the non-linear optimization problem into a sequence of tractable weighted least squares (WLS) problems, which do have analytical solutions.








```python
# Fit GLM
# Fit GLM (robust IRLS with tiny ridge to avoid singular matrices)
y_arr = np.asarray(y).astype(float).ravel()

def glm_binomial_irls_reg(X, y, max_iter=100, tol=1e-8, reg=1e-6):
    n_samples, n_features = X.shape
    beta = np.zeros(n_features)
    eps = 1e-9
    I = np.eye(n_features)
    for i in range(max_iter):
        # Compute the Linear Predictor
        eta = X @ beta
        mu = inv_logit(eta)
        mu = np.clip(mu, eps, 1 - eps)           # avoid zeros/ones
        # Calculate the observation weights
        W = mu * (1 - mu)
        # Compute the Adjusted Response Variable
        z = eta + (y - mu) / W
        # Weighted least squares using normal equations with ridge       
        XT_W = X.T * W  # broadcasting: (n_features, n_samples)
        A = XT_W @ X + reg * I
        b = XT_W @ z
        try:
            beta_new = np.linalg.solve(A, b)
        except np.linalg.LinAlgError:
            beta_new = np.linalg.lstsq(A, b, rcond=None)[0]
        if np.linalg.norm(beta_new - beta) < tol:
            beta = beta_new
            print(f'itr run {i}, max_itr = {max_iter}')
            break
        beta = beta_new
    return beta


```

```python
beta = glm_binomial_irls_reg(X, y_arr, max_iter=100, tol=1e-6, reg=1e-6)

print(beta.shape)
# Predict probabilities
eta = X @ beta
probs = inv_logit(eta)
preds = (probs >= 0.5).astype(int)

# Accuracy
accuracy = (preds == y_arr).mean()
print(f'Final accuracy: {accuracy:.4f}')
print('Model coefficients:', beta)
print(beta.shape)
```
