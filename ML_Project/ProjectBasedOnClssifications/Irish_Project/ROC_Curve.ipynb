```python
def ROC_CUREVEX(Y_TEST_DATA,Y_SCORE_DATA,TARGET_CLASS_NAME):
	# Compute ROC curve and AUC for each class
	fpr = dict()
	tpr = dict()
	roc_auc = dict()
	for i in range(len(TARGET_CLASS_NAME)):
		fpr[i], tpr[i], _ = roc_curve(Y_TEST_DATA[:, i], Y_SCORE_DATA[:, i])
		roc_auc[i] = auc(fpr[i], tpr[i])
	
	# Plot ROC curves
	plt.figure(figsize=(8, 6))
	colors = ['blue', 'green', 'red']
	for i in range(len(TARGET_CLASS_NAME)):
		plt.plot(fpr[i], tpr[i], color=colors[i], lw=2,
				label=f'ROC curve for class {TARGET_CLASS_NAME[i]} (AUC = {roc_auc[i]:.2f})')
	
	plt.plot([0, 1], [0, 1], 'k--', lw=1)
	plt.xlim([0.0, 1.0])
	plt.ylim([0.0, 1.05])
	plt.xlabel('False Positive Rate')
	plt.ylabel('True Positive Rate')
	plt.title('Multiclass ROC Curve - Logistic Regression on Iris Dataset')
	plt.legend(loc="lower right")
	plt.grid()
	plt.show()
```

```python
path = os.path.join(os.getcwd(), 'Iris.csv')
df = pd.read_csv(path)
df.drop(['Id'], axis=1, inplace=True)
df.columns

X = df.drop(['Species'], axis=1, inplace=False)
y = df['Species'].map({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
```

```python
target_names = df['Species'].value_counts().index.tolist()
target_names
```

```python
y_bin = label_binarize(y, classes=[0, 1, 2])  # One-hot encode targets
n_classes = y_bin.shape[1]
# Train/test split
X_train, X_test, y_train, y_test = train_test_split(X, y_bin, test_size=0.3, random_state=42)
# Standardize features
```

### ROC Curve for models 
penalty='l2'

solver='lbfgs'

```python
# Train One-vs-Rest Logistic Regression
model_l2_lbfgs = OneVsRestClassifier(LogisticRegression(random_state=42,penalty='l2',solver='lbfgs',max_iter=200,tol=0.001,fit_intercept=True))
model_l2_lbfgs.fit(X_train, y_train)
y_score = model_l2_lbfgs.predict_proba(X_test)
```

```python
# Predict on test set
y_pred = model_l2_lbfgs.predict(X_test)
# Evaluate performance
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
# Perform 5-fold cross-validation
scores = cross_val_score(model_l2_lbfgs, X, y, cv=5, scoring='accuracy')
# Output results
print("Cross-validation scores:", scores)
print("Mean accuracy: {:.2f}".format(np.mean(scores)))
```

```python
ROC_CUREVEX(Y_TEST_DATA = y_test,
            Y_SCORE_DATA = y_score,
            TARGET_CLASS_NAME = target_names
            )
```

```python
# Perform 5-fold cross-validation
scores = cross_val_score(model_l1_liblinear, X, y, cv=5, scoring='accuracy')
# Output results
print("Cross-validation scores:", scores)
print("Mean accuracy: {:.2f}".format(np.mean(scores)))
```

```python
# Predict on test set
y_pred = model_l1_liblinear.predict(X_test)
# Evaluate performance
print("Accuracy:", accuracy_score(y_test, y_pred))
```
