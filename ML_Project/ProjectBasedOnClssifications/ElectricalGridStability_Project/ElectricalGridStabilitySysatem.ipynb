```python
path = os.path.join(os.getcwd(), 'GridStability.csv')
df = pd.read_csv(path)
```

```python
df.columns
```

```python
df['stabf'].value_counts()
X = df.drop(['stabf'],axis='columns')
X = sm.add_constant(X)
y = df['stabf'].map({'unstable':0,'stable':1})
```

```python
# Fit GLM with Gaussian family (linear regression)
lnk = sm.genmod.families.links.Log()
model = sm.GLM(y, X, family=sm.families.Gaussian(link=lnk))
result = model.fit()
# Summary
print(result.summary())
```
