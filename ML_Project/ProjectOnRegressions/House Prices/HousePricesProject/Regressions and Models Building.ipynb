#### Qus: Ridge regression, an L2 regularization method, is suitable for constructing predictive models and validating their performance ?

Ridge regression = regularization method → useful for model building + validation ?

Ridge regression is a regularization method, and it can absolutely be used both to create models and to validate them.

Ridge regression adds an L2 penalty to the ordinary least squares (OLS) cost function. Shrinks coefficients toward zero but never eliminates them completely. Best when predictors are highly correlated (multicollinearity) or when you want to reduce overfitting while keeping all variables in the model

```python
def RidgeRegressionsAnalysis(X_train_scaled, y_train, X_test_scaled, y_test,alphas,solvers):
    results = []
    coeff_dict = {}
    n = X_test_scaled.shape[0]   # number of observations in test set
    p = X_test_scaled.shape[1]   # number of predictors (after encoding)
    y_mean = np.mean(y_test)
    for alpha in alphas:
        coeff_dict[alpha] = {}
        for solver in solvers:
            ridge = Ridge(alpha=alpha, solver=solver, random_state=42)
            ridge.fit(X_train_scaled, y_train)
            # Combine intercept first, then coefficients
            # Reshape intercept to match dimensions of coefficients
            intercept_reshaped = np.array([ridge.intercept_]).reshape(1, -1)
            coeffs_with_intercept = np.concatenate((intercept_reshaped, ridge.coef_.reshape(1, -1)), axis=1)
            # Do not required const term
            coeff_dict[alpha][solver] = ridge.coef_.reshape(1, -1)
            #predict ridge regressions
            y_pred = ridge.predict(X_test_scaled)
            # Adjusted R²
            r2_adj = 1 - ((1 - r2_score(y_test, y_pred)) * (n - 1)) / (n - p - 1)
            
            # Regression Sum of Squares (SSreg)
            ss_reg = np.sum((y_pred - y_mean) ** 2)  
            # Mean Square Regression (MSr)
            msr = ss_reg / p            
            results.append({
                "Alpha": alpha,
                "Solver": solver,
                "R2": r2_score(y_test, y_pred),
                "Adj_R2": r2_adj,
                "MSE": mean_squared_error(y_test, y_pred),
                'MSr':msr
            })
    return results,coeff_dict
```

**I am not using 'lbfgs' solver**: The lbfgs solver is a quasi‑Newton optimization algorithm. For Ridge,  restricts lbfgs to cases where positive=True. positive=True means the model will constrain all coefficients to be non‑negative. This is useful in domains where negative coefficients don’t make sense (e.g., economics, biology). The lbfgs solver was implemented specifically for this constrained optimization problem.

#### Learning curve for Ridge regression to validate Models:

Visualize training vs validation scores against increasing training set sizes. See whether model is underfitting, overfitting, or well‑balanced


```python
def ridge_regressions_learning_curve(X_train_learn_curv_scaled, y_train_learn_curv,learn_curve_alpha,learn_curve_solver,learning_score,train_sizes):
    # Create subplot grid (rows = learn_curve_alpha, cols = learn_curve_solver)
    fig, axes = plt.subplots(len(learn_curve_alpha), len(learn_curve_solver), figsize=(18, 16), sharex=True, sharey=True)
    fig.suptitle("Learning Curves for Ridge Regression (R² Score)", fontsize=18)
    
    # Loop through each combination
    for i, alpha in enumerate(learn_curve_alpha):
        for j, solver in enumerate(learn_curve_solver):
            ridge = Ridge(alpha=alpha, solver=solver, random_state=42)
            train_sizes_out, train_scores, test_scores = learning_curve(
                ridge, X_train_learn_curv_scaled, y_train_learn_curv,
                cv=5, scoring=learning_score, train_sizes=train_sizes,
                shuffle=True, random_state=42
            )
            
            # Mean and std
            train_mean = np.mean(train_scores, axis=1)
            train_std  = np.std(train_scores, axis=1)
            test_mean  = np.mean(test_scores, axis=1)
            test_std   = np.std(test_scores, axis=1)
    
            # Plot in correct subplot
            ax = axes[i, j]
            ax.plot(train_sizes_out, train_mean, 'o-', color="blue", label="Train")
            ax.fill_between(train_sizes_out, train_mean-train_std, train_mean+train_std, alpha=0.2, color="blue")
            ax.plot(train_sizes_out, test_mean, 'o-', color="green", label="CV")
            ax.fill_between(train_sizes_out, test_mean-test_std, test_mean+test_std, alpha=0.2, color="green")
            
            ax.set_title(f"Alpha={alpha}, Solver={solver}")
            ax.grid(True)
            if i == len(learn_curve_alpha)-1:
                ax.set_xlabel("Training Size")
            if j == 0:
                ax.set_ylabel("R² Score")
    # Add legend once
    handles, labels = ax.get_legend_handles_labels()
    fig.legend(handles, labels, loc='lower center', ncol=2, fontsize=12)
    
    plt.tight_layout(rect=[0, 0.05, 1, 0.97])
    plt.show()
```

```python
# Parameters
lrn_curve_alpha = [10, 50, 150, 500]
lrn_curve_solver = ['svd', 'cholesky', 'lsqr']
lrn_score = 'r2'
train_sizes = np.linspace(0.1, 1.0, 10)

ridge_regressions_learning_curve(X_train_learn_curv_scaled = X_train_scaled_std_scal_single_line, 
                                 y_train_learn_curv = y_train_single_line_sol,
                                 learn_curve_alpha = lrn_curve_alpha,
                                 learn_curve_solver = lrn_curve_solver,
                                 learning_score = lrn_score,
                                 train_sizes = train_sizes
                                 )
```

```python
def ridge_regression_coffificent_path_plot(cofficient_dict):
    alphas_path_plot = sorted(cofficient_dict.keys())
    solvers_path_plot = list(next(iter(cofficient_dict.values())).keys())
    
    # Create subplot grid: rows = solvers_path_plot
    fig, axes = plt.subplots(len(solvers_path_plot), 1, figsize=(12, 18), sharex=True)
    
    for i, solver in enumerate(solvers_path_plot):
        # Collect coefficients across alphas_path_plot for this solver
        coefs = []
        for a in alphas_path_plot:
            coefs.append(coeff_dict[a][solver].flatten())  # flatten to 1D
        
        coefs = np.array(coefs)
        
        # Plot coefficient paths
        ax = axes[i]
        for j in range(coefs.shape[1]):
            ax.plot(alphas_path_plot, coefs[:, j], marker='o', label=f"Feature {j}" if i == 0 else "")
        
        ax.set_xscale("log")
        ax.set_ylabel("Coefficient Value")
        ax.set_title(f"Solver: {solver}")
        ax.grid(True)
    
    axes[-1].set_xlabel("Alpha (log scale)")
    plt.tight_layout(rect=[0, 0, 1, 0.97])
    plt.show()

ridge_regression_coffificent_path_plot(coeff_dict)
```

# Residual Analysis Workflow in Ridge Regression



#### 1. Residuals vs Predicted Plot

```python

def residuals_vs_predicted_plots(X_train_scaled,y_train_scaled,X_test_scaled,y_test,alphas,solvers):
    # Create subplot grid (rows = alphas, cols = solvers)
    fig, axes = plt.subplots(len(alphas), len(solvers), figsize=(15, 12), sharex=True, sharey=True)
    fig.suptitle("Residuals vs Predicted Values (Ridge Regression)", fontsize=18)
    
    for i, alpha in enumerate(alphas):
        for j, solver in enumerate(solvers):
            ridge = Ridge(alpha=alpha, solver=solver, random_state=42)
            ridge.fit(X_train_scaled, y_train_scaled)
            y_pred = ridge.predict(X_test_scaled)
            residuals = y_test - y_pred
    
            ax = axes[i, j]
            ax.scatter(y_pred, residuals, alpha=0.7, color="blue")
            ax.axhline(y=0, color="red", linestyle="--")
            ax.set_title(f"Alpha={alpha}, Solver={solver}")
            if i == len(alphas)-1:
                ax.set_xlabel("Predicted Values")
            if j == 0:
                ax.set_ylabel("Residuals")
            ax.grid(True)
    
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

# Parameters
alphas_res_pred_plot = [10, 50, 150, 500]
solver_res_pred_plot = ['svd', 'cholesky','lsqr']  # Ensure all solvers are valid

residuals_vs_predicted_plots(X_train_scaled = X_train_scaled_std_scal_single_line,
                             y_train_scaled = y_train_single_line_sol,
                             X_test_scaled  = X_test_scaled_std_scal_single_line,
                             y_test         = y_test_single_line_sol,
                             alphas         = alphas_res_pred_plot,
                             solvers         = solver_res_pred_plot  # Ensure solver values are valid
                             )
```

#### 2. Histogram / Density Plot of Residuals

```python
def residuals_hist_plot(X_train_scaled,y_train,X_test_scaled,y_test,alphas,solvers):
    # Create subplot grid (rows = alphas, cols = solvers)
    fig, axes = plt.subplots(len(alphas), len(solvers), figsize=(15, 12), sharex=False, sharey=False)
    fig.suptitle("Histogram + KDE of Residuals (Ridge Regression)", fontsize=18)
    
    for i, alpha in enumerate(alphas):
        for j, solver in enumerate(solvers):
            ridge = Ridge(alpha=alpha, solver=solver, random_state=42)
            ridge.fit(X_train_scaled, y_train)
            y_pred = ridge.predict(X_test_scaled)
            residuals = y_test - y_pred
    
            ax = axes[i, j]
            sns.histplot(residuals, kde=True, bins=30, ax=ax, color="skyblue")
            ax.set_title(f"Alpha={alpha}, Solver={solver}")
            ax.set_xlabel("Residuals")
            ax.set_ylabel("Frequency")
            ax.grid(True)
    
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

# Parameters
alphas_res_pred_plot = [10, 50, 150, 500]
solver_res_pred_plot = ['svd', 'cholesky','lsqr']  # Ensure all solvers are valid

residuals_hist_plot(X_train_scaled = X_train_scaled_std_scal_single_line,
                    y_train = y_train_single_line_sol,
                    X_test_scaled = X_test_scaled_std_scal_single_line,
                    y_test = y_test_single_line_sol,
                    alphas = alphas_res_pred_plot,
                    solvers = solver_res_pred_plot
                    )
```

```python
def residuals_q_q_plot(X_train_scaled,y_train,X_test_scaled,y_test,alphas,solvers):
    # Create subplot grid (rows = alphas, cols = solvers)
    fig, axes = plt.subplots(len(alphas), len(solvers), figsize=(15, 12))
    fig.suptitle("Q-Q Plots of Residuals (Ridge Regression)", fontsize=18)
    
    for i, alpha in enumerate(alphas):
        for j, solver in enumerate(solvers):
            ridge = Ridge(alpha=alpha, solver=solver, random_state=42)
            
            ridge.fit(X_train_scaled_std_scal_single_line, y_train_single_line_sol)
            y_pred = ridge.predict(X_test_scaled_std_scal_single_line)
            residuals = y_test_single_line_sol.values.flatten() - y_pred.flatten()
        
            ax = axes[i, j]
            stats.probplot(residuals, dist="norm", plot=ax)
            ax.get_lines()[1].set_color('red')  # make Q-Q line red
            ax.set_title(f"Alpha={alpha}, Solver={solver}")

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

# Parameters
alphas_res_pred_plot = [10, 50, 150, 500]
solver_res_pred_plot = ['svd', 'cholesky','lsqr']  # Ensure all solvers are valid

residuals_q_q_plot(X_train_scaled = X_train_scaled_std_scal_single_line,
                    y_train = y_train_single_line_sol,
                    X_test_scaled = X_test_scaled_std_scal_single_line,
                    y_test = y_test_single_line_sol,
                    alphas = alphas_res_pred_plot,
                    solvers = solver_res_pred_plot
                    )
```

```python
def residuals_hist_plot(X_train_scaled,y_train,X_test_scaled,y_test,alphas,solvers,num_features):
    
    # Create subplot grid (rows = alphas, cols = solvers)
    fig, axes = plt.subplots(len(alphas), len(solvers), figsize=(18, 14), sharex=False, sharey=False)
    fig.suptitle("Residuals vs Features (First 6) - Ridge Regression", fontsize=18)
    
    for i, alpha in enumerate(alphas):
        for j, solver in enumerate(solvers):
            ridge = Ridge(alpha=alpha, solver=solver, random_state=42)
            ridge.fit(X_train_scaled, y_train)
            y_pred = ridge.predict(X_test_scaled)
            residuals = y_test - y_pred
    
            # Plot residuals vs first 6 features
            ax = axes[i, j]
            for f in range(num_features):
                ax.scatter(X_test_scaled[:, f], residuals, alpha=0.6, label=f"Feature {f}")
            ax.axhline(y=0, color="red", linestyle="--")
            ax.set_title(f"Alpha={alpha}, Solver={solver}")
            ax.set_xlabel("Feature values (scaled)")
            ax.set_ylabel("Residuals")
            ax.legend(loc="upper right", fontsize=8)
            ax.grid(True)
    
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

# Parameters
alphas = [10, 50, 150, 500]
solvers = ['svd', 'cholesky', 'lsqr']
num_features = 6  # first 6 features

residuals_hist_plot(X_train_scaled = X_train_scaled_std_scal_single_line,
                    y_train = y_train_single_line_sol,
                    X_test_scaled = X_test_scaled_std_scal_single_line,
                    y_test = y_test_single_line_sol,
                    alphas = alphas_res_pred_plot,
                    solvers = solver_res_pred_plot,
                    num_features = num_features)
```
