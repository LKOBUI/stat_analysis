1. Order of the Model

It is important to keep the order of the model as low as possible.

When the response function appears to be curvilinear, transformations should be tried to keep the model first order. If this fails, a second-order polynomial should be tried. Arbitrary fitting of high-order polynomials is a serious abuse of regression analysis.

2. Model-Building Strategy

Various strategies for choosing, One approach is to successively fit models of increasing order until the t test for the highest order term is nonsignificant.

Procedure (forward selection and backward elimination):
fit the highest order model and then delete terms one at a time, starting with the highest order, until the highest order remaining term has a significant t statistic.

3. Extrapolation:

Extrapolation with polynomial models can be extremely hazardous.
In general, polynomial models may turn in unanticipated and inappropriate directions, both in interpolation and in extrapolation.



4. Ill-Conditioning (I)

As the order of the polynomial increases, the X′X matrix becomes ill-conditioned. This means that the ***matrix inversion calculations*** will be inaccurate, and considerable error may be introduced into the parameter estimates.

Design matrix becomes numerically unstable.

If the values of x are limited to a narrow range, there can be significant ill-conditioning or multicollinearity in the columns of the X matrix.

**Spline interpolation:**

Spline interpolation is an alternative approach to data interpolation. Compare to ***polynomial interpolation using on single formula to correlate all the data points***, spline interpolation uses several formulas; each formula is a low degree polynomial to pass through all the data points. These resulting functions are called splines.

**Cubic Spline Interpolation:**

**The goal of cubic spline interpolation** is to get an interpolation formula that is **continuous in both the first and second derivatives**, both within the intervals and at the interpolating nodes. This will give us a smoother interpolating function. The continuity of first derivative means that ***the graph y = S(x) will not have sharp corner***s. The continuity of second derivative means that the radius of curvature is defined at each point.

### what is radius of curvature ?

The radius of curvature is a measure of how sharply a curve bends.
-	A smaller radius of curvature indicates a sharper, more pronounced bend.
-	A larger radius indicates a gentler curve.
-	A straight line, which has no curvature, has an infinite radius of curvature.

#### Choosing the number of knots:

The number of knots is often more critical than their precise location. 
-	Too few knots results in a stiff, underfitted model that fails to capture complex patterns in the data.
-	Too many knots (or knots placed at every data point) can lead to an overly flexible, "wiggly" model that overfits to the noise in the data. 

### Common strategies for selecting the number of knots include:

-	Information criteria: Use metrics like the Akaike Information Criterion (AIC) or the Bayesian Information Criterion (BIC) to select the optimal number of knots. You fit models with a range of different knot counts and choose the one that minimizes the criterion score, which balances goodness-of-fit with model complexity.
-	Cross-validation: Use a technique like k-fold cross-validation to assess the model's performance on unseen data for different numbers of knots. The knot count that yields the best performance (e.g., lowest mean squared error) is chosen.
-	Rules of thumb: For moderate sample sizes, a small number of knots (typically between 3 and 10) is often adequate.
-	Data-driven algorithms: More advanced methods like Generalized Additive Models (GAMs) and penalized splines (P-splines) automate the selection of the number of knots by using regularization. 

### Determining knot positions

The location of the knots can significantly influence the fit, though it is generally less critical than the number of knots. 

-	Quantile-based placement: The most standard and robust method is to place knots at equally-spaced quantiles of the independent variable. This ensures that there is roughly an equal number of observations between each knot, preventing overfitting in data-sparse regions.
-	Adaptive knots: Algorithms can be used to place knots adaptively, concentrating them in areas where the data changes rapidly. This is a more complex approach but can improve model performance.
-	Manual specification: Domain knowledge can inform knot placement. For example, if you know a physical process changes at a certain temperature, you can place a knot at that specific value.
-	Equally spaced knots: Placing knots at equally spaced intervals is a simpler alternative, but it can lead to problems if data is clustered in some regions and sparse in others.

### Qus: Explained Cubic Spline Models no continuity restrictions?

A **standard cubic spline**, by definition, enforces continuity of the function, its first derivative, and its second derivative at the knot points. To build a cubic model with no continuity restrictions, you are not building a spline. Instead, you are fitting a set of independent cubic polynomials, one for each defined interval of your data. This is typically referred to as piecewise cubic regression or segmented cubic regression.

1. No Continuity Restrictions

Definition

Each spline segment between knots is fit independently using a cubic polynomial. There's no requirement for the value, slope, or curvature to match at the knots.

Implication

The resulting function may be discontinuous at the knots. This means the curve can exhibit jumps or abrupt changes in slope or curvature at those points.

### Use Case

Rarely applied in practice, since most real-world phenomena are expected to be at least continuous.

2. Continuity Restrictions

Definition

At each knot in the spline:

-	The function is continuous (no jumps in value).
-	Typically, the first derivative (slope) and second derivative (curvature) are also continuous.

Implication

The fitted curve transitions smoothly at the knots. There are no abrupt jumps or sharp changes—each polynomial segment blends seamlessly into the next.

### Use Case

This is the standard method in cubic spline regression, as it yields a smooth and realistic fit for most real-world data

### Model Comparison Using Splines [Performed null hypothesis]- Method Involve compare models between Simple Linear or polynomials without spline & polynomials with cubical spline:

1. Fit the Models
-	**Null Model**: Fit a simpler model, such as linear regression or polynomial regression without spline terms.
-	***Alternative Model:*** Fit a cubic spline regression model, including knots and continuity restrictions.
2. Specify the Hypotheses
-	***Null Hypothesis (H₀):*** Spline terms do not improve the model fit; the simpler model is sufficient.
-	***Alternative Hypothesis (H₁):*** Spline terms significantly improve the model fit.

3. Perform the Test
A. Likelihood Ratio Test (LRT) / F-test
This is the most common approach for comparing models:
   1. Fit both models
       - Obtain their Residual Sum of Squares (RSS) or log-likelihoods (LL).
   2. Calculate the test statistic
       - F-test (using RSS):

       
       Degree of freedom should be the residuals

4.	Compare the test statistic
   -	Use the F-distribution for the F-test
   -	Use the chi-square distribution for the LRT
   - Compute the p-value to assess statistical significance.

B. Wald Test (for Individual Spline Coefficients)

-	Test whether each spline coefficient is significantly different from zero.
-	Use standard errors from the regression output to compute t-statistics and p-values.
-	The Wald test is a statistical test used to assess the significance of individual coefficients (or sets of coefficients) in regression models, including cubic spline polynomial regression.

Wald Test in Cubic Spline Regression
Context

-	In cubic spline regression, the model includes several coefficients:
    - Those for the base polynomial
    - Those for the spline basis functions (which capture the flexibility at and around the knots)
-	You may want to test:
    - Whether a single spline coefficient is significantly different from zero
    - Whether a group of spline coefficients (e.g., all associated with the spline terms) are jointly significan







Interpretation

-	***Significant Wald test (p < 0.05):***
   - The coefficient(s) are significantly different from zero; the spline terms contribute to the model.
-	***Not significant:***

-	If p-value < significance level (e.g., 0.05):
   - → Reject the null hypothesis (H₀); spline terms significantly improve the model fit.
-	If p-value ≥ significance level:
   - → Fail to reject H₀; the simpler model is sufficient.

### 7.3 NONPARAMETRIC REGRESSION

Nonparametric Regression:[ without assuming a predetermined shape]
 
Nonparametric regression is a type of regression analysis that ***makes few or no assumptions*** about the functional form of the relationship between the independent variable(s) and the dependent variable. Instead of specifying a fixed equation (like a straight line or a polynomial), nonparametric regression lets the data "speak for itself" and can capture complex, nonlinear relationships.


Key Features of Nonparametric Regression:

1. No Fixed Functional Form
   -	Unlike parametric regression (e.g., linear regression: 
y = β0 + β1x), nonparametric regression does not assume a specific equation for the relationship.
   -	The shape of the regression curve is determined by the data.

2. Examples of Nonparametric Regression Methods
-	Kernel regression (Nadaraya–Watson estimator)
-	Local polynomial regression (LOESS/LOWESS)
-	Spline regression (including cubic splines)
-	Nearest neighbor regression
-	Regression trees (e.g., CART, random forests)

Pros: Very flexible; can fit almost any shape.

Cons: May require more data, can be sensitive to noise, and may overfit if not properly regularized.

The basic idea of nonparametric regression is to develop a model-free basis for predicting the response over the range of the data.

Most of the nonparametric regression literature focuses on a single regressor; however, many of the basic ideas extend to more than one.

### 7.3.1 Kernel Regression

Kernel Regression

Introduction

Kernel regression is a popular nonparametric regression technique that estimates the relationship between variables without assuming a specific functional form. It’s especially useful for uncovering complex, nonlinear relationships in data.

### What is Kernel Regression?

Kernel regression estimates the value of the dependent variable y at a given point x by taking a weighted average of the observed y values.

The weights depend on how close each observed xi is to the target x, and are determined by a kernel function.









### Bandwidth Selection

•	Small bandwidth:
- Less smoothing; the curve closely follows the data.
- Risk of overfitting.

•	Large bandwidth:
- More smoothing; the curve becomes smoother.
- Risk of underfitting.

•	**Bandwidth selection is crucial**and is often performed using cross-validation.

### Kernel smoothers use a bandwidth, b matrix:

### Property of weight:

Weights are chosen such that wij ≅ 0 for all yi’s outside of a defined “neighborhood” of the specific location of interest.

A large value for b results in more of the data being used to predict the response at the specific location.

b decreases, less of the data are used to generate the prediction, and the resulting plot looks more “wiggly” or bumpy.

### When to use parametric model

When it provides a reasonable and satisfactory fit to the data. The complexity issue is not trivial. Simple models provide an easy and convenient basis for prediction. In addition, the model terms
