### What is Basis Spline? 

Basis spline, or B-spline, is a piecewise polynomial function that is used for smoothing data.

It is defined by a set of control points and a degree, which determines the polynomial degree of the segments between the control points. ***The B-spline is constructed using a set of basis functions, which are defined over a set of knots. The B-spline is a linear combination of these basis functions, where the coefficients are the weights assigned to each basis function***

***B-splines (Basis splines)*** are a family of piecewise polynomial functions that are widely used in curve fitting, computer graphics, and nonparametric regression. They provide a **flexible** and ***numerically stable*** way to represent smooth curves




3. Knot Vector
-	The knot vector determines where and how the basis functions change.
-	For a cubic B-spline (k = 4), the basis functions are piecewise cubic polynomials.



5. Properties of B-splines
-	Local control: Changing one control point only affects the curve locally.
-	Smoothness: B-splines are C^{k-2} continuous (for degree k - 1), meaning they are smooth up to the (k - 2)-th derivative.
-	Non-negativity: All basis functions are non-negative.
-	Partition of unity: The sum of all basis functions at any x is 1.



### In Summary:

-	B-spline curves are smooth, flexible, and locally controlled curves defined as a sum of basis functions.
-	The basis functions are defined recursively and depend on the knot vector and the degree.
-	Cubic B-splines (degree 3) are most commonly used for smooth curve fitting.
