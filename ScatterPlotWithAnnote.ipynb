```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
x = np.random.randn(200)
y = np.random.randn(200)

# Example 1a: highlight by index list
highlight_idx = [2, 5, 40, 150]
RenameHilightedIdx = ['A','B','C','D']

plt.figure(figsize=(7,5))
plt.scatter(x, y, color='lightgray', alpha=0.6, label='others')
plt.scatter(x[highlight_idx], y[highlight_idx], color='red', s=80, edgecolor='k', label='highlighted')
for idx,i in zip(RenameHilightedIdx,highlight_idx):
    plt.annotate(f'{i}'+idx, (x[i], y[i]), 
                 xytext=(x[i] + 40, y[i] + 40),     # text location
                 arrowprops=dict(arrowstyle='->', color='black', lw=1.5), 
                 textcoords='offset points', 
                 bbox=dict(boxstyle='round,pad=0.3', fc='yellow', alpha=0.8)
                 ,fontsize=9)
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.title('Scatter with highlighted indices')
plt.show()

# Example 1b: highlight by condition (e.g., y > 1)
cond = y > 1.0
plt.figure(figsize=(7,5))
plt.scatter(x[~cond], y[~cond], color='lightgray', alpha=0.5)
plt.scatter(x[cond], y[cond], color='C1', s=70, edgecolor='k')
plt.title('Scatter with condition-based highlights (y > 1.0)')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
x = np.random.randn(100)
y = np.random.randn(100)

# choose a point to highlight
idx = 27
px, py = x[idx], y[idx]

plt.figure(figsize=(7,5))
plt.scatter(x, y, color='lightgray', alpha=0.6)
plt.scatter([px], [py], color='red', s=80, edgecolor='k', zorder=5)

# Arrow + text annotation pointing to the point
plt.annotate(
    'Important point',               # text
    xy=(px, py),                     # point to annotate
    xytext=(px + 0.8, py + 0.6),     # text location
    arrowprops=dict(arrowstyle='->', color='black', lw=1.5),
    bbox=dict(boxstyle='round,pad=0.3', fc='yellow', alpha=0.8),
    fontsize=10
)

plt.xlabel('x'); plt.ylabel('y'); plt.title('Scatter with arrow annotation')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("penguins").dropna()
ax = sns.scatterplot(data=df, x="bill_length_mm", y="bill_depth_mm", hue="species", palette="deep", s=60)

# choose a point to highlight (index or condition)
idx = df["bill_length_mm"].idxmax()
x, y = df.loc[idx, "bill_length_mm"], df.loc[idx, "bill_depth_mm"]

# marker + annotation with arrow
ax.scatter([x], [y], color="red", s=120, edgecolor="k", zorder=5)
ax.annotate(
    "Longest bill",
    xy=(x, y),
    xytext=(x + 5, y + 2),
    arrowprops=dict(arrowstyle="->", color="black", lw=1.2),
    bbox=dict(boxstyle="round,pad=0.3", fc="yellow", alpha=0.8),
    fontsize=10
)

ax.set_title("Penguins: bill length vs depth")
plt.show()
```
