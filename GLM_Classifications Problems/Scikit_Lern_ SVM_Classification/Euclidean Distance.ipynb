**Euclidean Distance Formula**

The Euclidean distance between two points (x1, y1) and (x2, y2) in 2D space is given by:

√((x2 - x1)^2 + (y2 - y1)^2)

**General Formula**

The Euclidean distance between two points x = (x1, x2, ..., xn) and y = (y1, y2, ..., yn) in n-dimensional space is given by:

√((x1 - y1)^2 + (x2 - y2)^2 + ... + (xn - yn)^2)


```python
point1 = np.array([1, 2, 3])
point2 = np.array([4, 5, 6])

# Calculate the Euclidean distance
distance = np.linalg.norm(point1 - point2)

print(distance)
```

***Vector Norms ||X||:***

A vector norm, sometimes represented with a double bar as ∥x∥, is a function that assigns a non-negative length or size to a vector x in n-dimensional space. Norms are essential in mathematics and machine learning for measuring vector magnitudes and calculating distances.












### Uses of numpy.linalg.norm
