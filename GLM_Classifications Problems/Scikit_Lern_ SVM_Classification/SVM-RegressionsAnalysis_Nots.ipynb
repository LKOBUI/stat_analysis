### Qus: What happing in ε-SVM regression?

In ε-SVM regression, the set of training data includes predictor variables and observed response values. The goal is to find a function f(x) that deviates from yn by a value no greater than ε for each training point x, and at the same time is as flat as possible.

### Duality (optimization): LINK

In mathematical optimization theory, **duality** or the **duality principle** is the principle that optimization problems may be viewed from either of two perspectives, the **primal problem** or the dual problem.

If the primal is a minimization problem then the dual is a maximization problem (and vice versa).

Any feasible solution to the primal (minimization) problem is at least as large as any feasible solution to the dual (maximization) problem. Therefore, the solution to the primal is an upper bound to the solution of the dual, and the solution of the dual is a lower bound to the solution of the primal.

#### Solving the SVM Regression Optimization Problem

#### Solver Algorithms

Prerequisite

1.	quadratic programming algorithms
2.	Gram matrix 
3.	Decomposition methods (also called chunking and working set methods)
4.	Sequential minimal optimization (SMO)

Use **standard quadratic programming** form and solved using common quadratic programming techniques to solve SVM regressions problem. But problem with these solutions it can be computationally expensive to use quadratic programming algorithms, especially since the Gram matrix may be too large to be stored in memory.

***Solutions use Decomposition methods***

Decomposition methods (also called chunking and working set methods) separate all observations into two disjoint sets: the working set and the remaining set. 

### Qus: How decomposition method reduce amount of storage?

A decomposition method modifies only the elements in the working set in each iteration. Therefore, only some columns of the Gram 

matrix are needed in each iteration, which reduces the amount of storage needed for each iteration.

### Sequential minimal optimization (SMO): Another popular method:

**Sequential minimal optimization (SMO)** is the most popular approach for solving SVM problems. SMO performs a series of two-point optimizations. In each iteration, a working set of two points are chosen based on a selection rule that uses second-order information. Then the Lagrange multipliers for this working set are solved analytically.
