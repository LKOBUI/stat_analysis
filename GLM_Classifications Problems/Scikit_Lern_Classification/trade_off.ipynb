**Precision**

Precision is the ratio of true positives to the sum of true positives and false positives. It measures the accuracy of the model's positive predictions.

**Recall**

Recall is the ratio of true positives to the sum of true positives and false negatives. It measures the model's ability to detect all instances of the positive class.

**Trade-off**

The precision/recall trade-off arises because increasing precision often decreases recall, and vice versa. This is because:

- Increasing precision requires reducing false positives, which may lead to more false negatives and decreased recall.
- Increasing recall requires reducing false negatives, which may lead to more false positives and decreased precision.
