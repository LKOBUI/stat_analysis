


Calculate Accuracy: (TP + TN) / (TP + TN + FP + FN). 

Calculate Precision: TP / (TP + FP). 

Calculate Recall: TP / (TP + FN). 

- **Precision** measures the accuracy of positive predictions.

- **Recall (Sentivity)** measures the model's ability to find all the positive instances.  

- **F1 Score** is the harmonic mean of precision and recall, providing a single score that balances both 
metrics.

   - **What it tells you**: This is particularly useful for imbalanced datasets, where accuracy can be    
   misleading. A high F1 Score indicates a good balance between precision and recall
