### Power law:

A power law in statistics describes a relationship between two quantities where one varies as a power of the other. It’s a type of scaling law and is often used to model phenomena with heavy tails or extreme events.
The relationship is expressed as 

y = ax^k

where 'x' and 'y' are quantities, and 'a' and 'k' are constants.



A signature characteristic of a power law is its scale invariance, meaning a relative change in 'x' results in a proportional relative change in 'y' that is independent of their initial values. A common example is the Pareto principle or "80/20 rule," where 80% of wealth is held by 20% of the population, illustrating how a small number of events or individuals can dominate the overall distribution.

### Scale Invariance:

If you multiply 'x' by a constant 'c', 'y' will only be scaled by a factor of c^k, demonstrating that the relationship doesn't depend on the absolute size of the quantities.





### Qus: Why Logarithmic transformation?

When a log transformation is applied to non-normal distribution, it tries to expand the differences between the smaller values because the slope for the logarithmic function is steeper for smaller values whereas the differences between the larger values can be reduced because, for large values, log distribution has a moderate slope. 




### Examples in Real-World Phenomena:

#### Natural Disasters:

The Gutenberg-Richter law describes the power-law distribution of earthquake magnitudes, where larger quakes are much less frequent than smaller ones. 

#### Economics:

The Pareto distribution (the "80/20 rule") is a power law that shows how wealth and income are distributed, with a small percentage of the population holding a large portion of the wealth.

#### Biology:

Power laws can govern the frequency of visits to different patches of surface by bacteria, with a few areas receiving most of the attention. 

#### Network Science:

The number of connections in a network often follows a power law distribution, indicating that a few highly connected nodes dominate the network.
