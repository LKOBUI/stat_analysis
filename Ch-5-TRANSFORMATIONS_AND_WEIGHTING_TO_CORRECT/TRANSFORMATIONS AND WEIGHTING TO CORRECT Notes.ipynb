### VARIANCE-STABILIZING TRANSFORMATIONS

### Qus: Why we required this, Method?

Ans: When variance not const. So we need to stabilized this variance.

### Qus: In which situations variance not const? And how you know if variance not const.

Ans: Let suppose response variable dependent upon follow the probability distributions in which the variance functionally related to mean. 

For example, if response variable y is a Poisson random variable in a simple source. So variance of y is equal to the mean.

Since the mean of y is related to the regressor variable x, the variance of y will be proportional to x.
So this is how variance of response variable not const.

### Qus: How to avoided this?

We can stabilizing variance by Useful Variance-Stabilizing Transformations technique.
