### Entropy:

Entropy is a measure of uncertainty or randomness in a system.

Types of Entropy
1. Shannon Entropy: Measures the uncertainty in a probability distribution.
2. Thermodynamic Entropy: Measures the disorder or randomness in a physical system.

0 ≤ H(X) ≤ log2(K)

- Minimum entropy (0): Occurs when one outcome has a probability of 1, and all other outcomes have a probability of 0.
                       Very less uncertainty or randomness in data or called Data biased.
- Maximum entropy (log2(K)): Occurs when all outcomes are equally likely. Very high uncertainty present in data set

### Gini Impurity

Gini impurity measures the probability of misclassifying a randomly chosen element from a node.

**Qus How Gini Differ from Entropy ?**

Entropy measures the amount of uncertainty or randomness in a probability distribution.

**Comparison**
1. Computational efficiency: Gini impurity is generally faster to compute than entropy.
2. Sensitivity to class distribution: Entropy is more sensitive to changes in class distribution than Gini impurity.


**Choosing Between Gini Impurity and Entropy**

1. Default choice: Both Gini impurity and entropy can be effective, and the choice often depends on the specific problem.
2. Consider computational efficiency: If computational efficiency is a concern, Gini impurity might be preferred.

### Hyperparameter:

Qus: Why need to Regularization Hyperparameters?

To avoided overfitting and underfitting. 

Overfitting occurs with non-parametric models & underfitting ouncers with parametric model.

**Qus: How to avoide over or underfitting?** 

Decision Trees make very few assumptions about the training data (as opposed to linear models, which obviously assume that the data is linear, for example). If left unconstrained, the tree structure will adapt itself to the training data, fitting it very closely, and most likely overfitting it. Such a model is often called a nonparametric model, not because it does not have any parameters (it often has a lot) but because the number of parameters is not determined prior to training, so the model structure is free to stick closely to the data. In contrast, a parametric model such as a linear model has a predetermined number of parameters, so its degree of freedom is limited, reducing the risk of overfitting (but increasing the risk of underfitting).

To avoid overfitting the training data, you need to restrict the Decision Tree’s freedom during training. As you know by now, this is called regularization.



### Regularization Hyperparameters in ScikitLern API

1. max_depth
2. min_samples_split (the minimum number of samples a node must have before it can be split)
3. min_samples_leaf (the minimum number of samples a leaf node must have)
4. min_weight_fraction_leaf (same as min_samples_leaf but expressed as a fraction of the total number of weighted instances)
5. max_leaf_nodes (maximum number of leaf nodes), 
6. max_features (maximum number of features that are evaluated for splitting at each node).

#### **Thumb Rule**

**Increasing** min_* hyperparameters or **reducing** max_* hyperparameters will regularize the
model.

### Another method of regularizations hyperparameters (pruning (deleting) unnecessary nodes)

A node whose children are all **leaf nodes is considered unnecessary** if the purity improvement it provides is **not statistically significant**. 

Standard statistical tests, such as the **χ2 test**, are used to estimate the probability that the improvement is purely the result of chance (which is called the **null hypothesis**). 

If this probability, called the **pvalue**, is **higher than a given threshold** (typically 5%, controlled by a hyperparameter), then the **node is considered unnecessary** and its **children are deleted**. The pruning continues until all unnecessary nodes have been pruned.

### The CART Training Algorithm:

Scikit-Learn uses the Classification And Regression Tree (CART) algorithm to train
Decision Trees (also called “growing” trees).

CART cost function for classification:



CART algorithm is a **greedy algorithm**: it greedily searches for an optimum split at the top level, then repeats the process at each level. It does not check whether or not the split will
lead to the lowest possible impurity several levels down. A greedy algorithm often produces a reasonably good solution, but it is not guaranteed to be the optimal solution.
