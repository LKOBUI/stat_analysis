### **L1 Regularization: Lasso Regression**
1. L1 regularization, or lasso (“least absolute shrinkage and selection operator”) regression, is 
   a regularization method that penalizes high-value coefficients in a machine learning model. 

2. Both penalty terms for L1 and L2 regularization are controlled by the model hyperparameter 
   lambda (λ), which determines the tradeoffs between bias and variance in coefficients.

   

3. Lasso regression can reduce coefficient values to zero. So we can use Lasso enabling feature 
   selection and removal. This aspect also allows lasso regression to handle some multicollinearity

4. lasso regression isn’t suitable for high multicollinearity — if there are highly-correlated 
   covariates present, L1 regularization will randomly remove one of the features from the model. 

5. Used to resolve overfitting. L1 regularization can actually cause model underfitting based on 
   the lambda value.
