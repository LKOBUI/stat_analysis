```python
data = {
    'Loves Popcorn': ['Yes', 'Yes', 'No', 'No', 'Yes', 'Yes', 'Yes', 'No'],
    'Loves Soda': ['Yes', 'No', 'Yes', 'Yes', 'Yes', 'No', 'No', 'No'],
    'Age': [7, 12, 18, 35, 38, 50, 83, None],
    'Loves Cool As Ice': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'No', 'No']
}

# Creating the DataFrame
df = pd.DataFrame(data)
df.dropna(inplace=True)
df, df.columns
```

```python
def gini(y):
    classes, counts = np.unique(y, return_counts=True)
    impurity = 1.0 - sum((count/len(y))**2 for count in counts)
    return impurity

def split_data(X, y, feature_index, threshold):
    left_mask = X[:, feature_index] <= threshold
    right_mask = X[:, feature_index] > threshold
    return left_mask, right_mask

def weighted_impurity(y_left, y_right):
    total = len(y_left) + len(y_right)
    return (len(y_left) / total) * gini(y_left) + (len(y_right) / total) * gini(y_right)
```

```python

def build_tree(X, y, depth=0, max_depth=5):
    # Check stopping criteria
    if len(np.unique(y)) == 1 or depth >= max_depth:
        # Return leaf with most common class
        return np.bincount(y).argmax()

    best_feature, best_thresh, min_impurity = None, None, float('inf')
    for feature_index in range(X.shape[1]):
        thresholds = np.unique(X[:, feature_index])
        for threshold in thresholds:
            left_mask, right_mask = split_data(X, y, feature_index, threshold)
            if len(y[left_mask]) == 0 or len(y[right_mask]) == 0:
                continue
            impurity = weighted_impurity(y[left_mask], y[right_mask])
            if impurity < min_impurity:
                min_impurity = impurity
                best_feature = feature_index
                best_thresh = threshold

    if best_feature is None:
        # No valid split found
        return np.bincount(y).argmax()

    # Create subtrees
    left_mask, right_mask = split_data(X, y, best_feature, best_thresh)
    left_subtree = build_tree(X[left_mask], y[left_mask], depth+1, max_depth)
    right_subtree = build_tree(X[right_mask], y[right_mask], depth+1, max_depth)

    return {
        'feature_index': best_feature,
        'threshold': best_thresh,
        'left': left_subtree,
        'right': right_subtree
    }
```
