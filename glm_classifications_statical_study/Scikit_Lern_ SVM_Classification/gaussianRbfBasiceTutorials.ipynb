#### What is Kernel Function?

Kernel Function is used to transform n-dimensional input to m-dimensional input, where m is much higher than n then find the dot product in higher dimensional efficiently. The main idea to use kernel is: A linear classifier or regression curve in higher dimensions becomes a Non-linear classifier or regression curve in lower dimensions.

#### The Gaussian RBF (Radial Basis Function):

What is a Kernel?

A kernel is a function that ***computes the similarity (or inner product) between two data points*** in a (possibly) ***higher-dimensional space***, ***without explicitly transforming the data***. This is called the ***kernel trick***."



#### Intuition:

- The kernel value is **close to 1** if x and x' are very similar (distance is small).
- The kernel value is **close to 0** if x and x' are very different (distance is large).
- It measures how "close" two points are in the feature space.

#### Why Use the Gaussian RBF Kernel?

- **Non-linear separation**: It allows SVMs and other algorithms to find non-linear decision boundaries.
- **Flexibility**: Can model complex relationships in data.

#### Parameters

- σ or γ: Controls the smoothness of the decision boundary.
- Small σ (large γ): Each data point influences only its close neighbors (can lead to overfitting).
- Large σ (small γ): Each data point influences a larger region (can lead to underfitting)



#### What is kernel trick in Gaussian RBF kernels ?

The kernel trick is a powerful concept in machine learning, especially in algorithms like Support Vector Machines (SVMs), that allows you to **work with high-dimensional feature spaces without explicitly** computing the coordinates in those spaces.

#### Kernel Trick Explained (with Gaussian RBF Kernel)

1. The Problem:

- Many algorithms (like SVMs) work by **finding linear boundaries** in the feature space.
- Real-world data is often not linearly separable in its original space.
- You could **transform data into a higher-dimensional space where it becomes linearly separable**, but this is computationally expensive or even impossible.

2. The Solution: Kernel Trick

- Instead of **explicitly transforming data**, you use a **kernel function to compute the inner product** (similarity) between data points as if they were mapped into a higher-dimensional space.
- The algorithm only needs these **inner products**, not the actual coordinates.

3. Gaussian RBF Kernel and the Trick



- The Gaussian RBF kernel computes similarity between two points.
- This is equivalent to computing the dot product in an infinite-dimensional space (because the RBF kernel corresponds to mapping data into an infinite set of features).
- You never need to compute the mapping explicitly—just use the kernel function.

4. Why is This Useful?

- Efficiency: You avoid the computational cost of working in high or infinite dimensions.
- Power: You can fit complex, non-linear boundaries using algorithms designed for linear problems.



"The kernel trick lets you use the Gaussian RBF kernel to implicitly work in a high-dimensional space, enabling powerful non-linear modeling without ever computing the actual transformation."

#### Transforming Linear Algorithms into Infinite-dimensional Nonlinear Classifiers and Regressors

If we apply any of the algorithms like perceptron Algorithm or linear regression on RBF kernel, actually we would be applying our algorithm to new infinite-dimensional data point we have created. Hence it will give a hyperplane in infinite dimensions, which will give a very strong non-linear classifier or regression curve after returning to our original dimensions.



So, Although we are applying linear classifier/regression it will give a non-linear classifier or regression line, that will be a polynomial of infinite power. And being a polynomial of infinite power, Radial Basis kernel is a very powerful kernel, which can give a curve fitting any complex dataset

#### Data Transform to Higher degree with Gaussian functions

```python
# Define Gaussian
def gaussian_rbf(x, landmark, gamma):
    return np.exp(-gamma * np.linalg.norm(x - landmark, axis=1)**2)
```

```python
gamma = 0.3

x1s = np.linspace(-4.5, 4.5, 200).reshape(-1, 1)
x2s = gaussian_rbf(x1s, -2, gamma)
x3s = gaussian_rbf(x1s, 1, gamma)
```
