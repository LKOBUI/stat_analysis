```python
path = os.path.join(os.getcwd(), 'DataSet', 'ChallengerData.txt')
df = pd.read_csv(path)
print(df.columns)
#print(df)
formulas = 'AtLeastOne_O_RingFailure ~ TempAtLaunch_F'
y, X = dmatrices(
                 formula_like = formulas, 
                 data=df,
                 return_type='dataframe'
                 )
```

```python
res = sm.Logit(y, X).fit()
res.summary()
```

```python
glm_link = sm.families.links.Logit()
glm_families = sm.families.Binomial(link=glm_link)
model_logit = sm.GLM(y, X,
                     family=glm_families)
result_logit = model_logit.fit()
```

```python
print(result_logit.summary())
print(f'p value: {result_logit.
      pvalues}')
```
