### **Softmax Regression**

Support multiple classes directly. Called Softmax Regression, or Multinomial Logistic Regression.

#### The idea is quite simple: 

when given an instance x, the Softmax Regression model
first computes a score sk(x) for each class k, then estimates the probability of each
class by applying the softmax function (also called the normalized exponential) to the
scores.

#### Softmax score for class k



θ(k) : parameter vector 

it computes the exponential of every score, then normalizes them (dividing by the sum of all the exponentials). The scores are generally called logits or log-odds (although they are actually unnormalized logodds).

#### Softmax function:



• K is the number of classes.

• s(x) is a vector containing the scores of each class for the instance x.

• σ(s(x))k is the estimated probability that the instance x belongs to class k given
   the scores of each class for that instance.

The Softmax Regression classifier predicts the class with the highest estimated probability.

#### Softmax Regression classifier prediction:



The argmax operator returns the value of a variable that maximizes a function. In
this equation, it returns the value of k that maximizes the estimated probability
σ(s(x))k.



#### Cross entropy cost function:

Cross entropy originated from information theory.

Minimizing the cost function, called the cross entropy.

Cross entropy is frequently used to measure how well a set of estimated class probabilities match the target classes.



y(i)^k is the target probability that the ith instance belongs to class k. In general, it is
either equal to 1 or 0, depending on whether the instance belongs to the class or not.

#### Cross entropy gradient vector for class k



Use Gradient Descent (or any other optimization algorithm) to find the parameter matrix Θ that minimizes the cost function.

#### Softmax regressions by using LogisticRegression:

##### C Argument In Logistice ?

C  default=1.0
Inverse of regularization strength; must be a positive float. 

What is the inverse of regularization strength in Logistic Regression?

We know that Regularization is applying a penalty to increasing the magnitude of parameter values in order to reduce overfitting. 

The problem comes when you have a lot of parameters (a lot of independent variables) but not too much data. which means it fits your data almost perfectly i.e. overfittitng.

To solve this, as well as minimizing the error Most often the function is λΣθ(j)^2, which is some constant λ times the sum of the squared parameter values θ(j)^2.



The larger λ is the less likely it is that the parameters will be increased in magnitude simply to adjust for small perturbations in the data.

Avoide Larger λ

 C=1/λ

 Qus: How to select C ?

  1. The regularization strength is determined by C and as C increases, the regularization term becomes smaller. Extremely large C values, it's as if there is no regularization at all.

  If the initial model is overfit (as in, it fits the training data too well), then adding a strong regularization term (with small C value) makes the model perform worse for the training data, but introducing such "noise" improves the model's performance on unseen (or test) data.

#### Multinomial Logit

We can extend the classical logistic regression model to cases involving more than two categorical outcomes. First consider a case where there are m + 1 possible categorical outcomes, but the outcomes are nominal.
By this we mean that there is no natural ordering of the response categories. Let the outcomes be represented by 0,1,2,..., m. The probabilities that the responses on observation i take on one of the m + 1
possible outcomes can be modeled as



Notice that there are m parameter vectors. Comparing each response category to a baseline categroy produces ***logits***



where our choice of zero as the baseline category is arbitrary. Maximum likelihood estimation of the parameters in these models is fairly straightforward.

```python
X, y = make_classification(1000, 200, n_informative=195, random_state=2023)
# split into train-test datasets
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=2023)

# normalize the data
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

scores = {}
for C in (10**k for k in range(-6, 6)):
    lr = LogisticRegression(C=C)
    lr.fit(X_train, y_train)
    scores[C] = {'train accuracy': lr.score(X_train, y_train), 
                 'test accuracy': lr.score(X_test, y_test)}

# plot the accuracy scores for different values of C
pd.DataFrame.from_dict(scores, 'index').plot(logx=True, xlabel='C', ylabel='accuracy');  
    
```

```python

path = os.path.join(os.getcwd(), 'DataSet', 'Iris.csv')
df = pd.read_csv(path)
df['Species'].value_counts()
df.columns
```

```python
def getMap(x):
    if x == 'Iris-setosa':
        return 0
    if x == 'Iris-versicolor':
        return 1
    if x == 'Iris-virginica':
        return 2 

df['sps_nw'] = df['Species'].map(lambda x: getMap(x))
```

```python
X = df.loc[:,['PetalLengthCm','PetalWidthCm']].to_numpy()
X.shape
y = df['sps_nw'].to_numpy().reshape(X.shape[0],1)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
# applies ℓ2 regularization
softmax_reg = LogisticRegression(C=30, random_state=42)
softmax_reg.fit(X_train, y_train)
```

#### How to predict with above models:
Given patel length = 5 cm
      Petal width  = 2 cm

And ask your models what kind of this Iris


```python
Ires_TypeX = ['Iris-setosa','Iris-versicolor','Iris-virginica']
Ires_Type = softmax_reg.predict([[5, 2]])
print('predicted type Ires: ',Ires_TypeX[Ires_Type[0]])
print('predicted probility: ',softmax_reg.predict_proba([[5, 2]]).round(2))
```
