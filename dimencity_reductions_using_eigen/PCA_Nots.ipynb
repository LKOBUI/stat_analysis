### Explain about orthogonal ?

"Orthogonal" means "perpendicular". 

Two vectors are said to be orthogonal to each other if and only their dot product is zero. 
In an orthogonal matrix, every two rows and every two columns are orthogonal and the length of every row (vector) or column (vector) is 1.

Properties of Orthogonal Matrix:

-	Transpose and Inverse are equal. i.e., A-1 = AT.
-	The product of A and its transpose is an identity matrix. i.e., AAT = ATA = I.
-	An identity matrix (I) is orthogonal as I · I = I · I = I.

Determinant of Orthogonal Matrix

Determinant of an orthogonal matrix is +1 or -1.

det(A) = ±1

### What is meaning of transformation is orthogonal?

In the context of linear algebra, a transformation is "orthogonal" if it preserves the dot product between vectors, meaning it doesn't change the length or angle between vectors. This also implies that the transformation preserves the Euclidean norm of a vector. 

### Properties of an orthogonal transformation:





Examples of orthogonal transformations:

-	Rotations: Rotating a vector around a specific axis is an orthogonal transformation. 
-	Reflections: Reflecting a vector across a line or plane is an orthogonal transformation. 
-	Compositions of rotations and reflections: Combinations of these transformations are also orthogonal.

### Qus: Explain stationary points of differential functions.

A stationary point of a differential function is a point where the function's derivative (or gradient) is equal to zero. These points are often called "turning points" or "critical points" because the function changes direction at these points, potentially going from increasing to decreasing or vice versa. Graphically, this corresponds to points on the graph of f(x)f(x) where the tangent to the curve is a horizontal line.
